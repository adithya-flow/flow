# Dictionaries

A dictionary is a mutable collection of key-value pairs.

Dictionaries are written using curly braces:

```python
student = {
    "name": "Alice",
    "age": 21,
    "score": 92
}
```

Each key maps to a value.

Dictionaries are useful when data has meaningful labels rather than only positional indexes.

```python
student["name"]   # "Alice"
student["score"]  # 92
```

## Important properties

- Dictionaries are mutable.
- Keys must be hashable.
- Values can be of almost any Python type.
- Keys are unique.
- In modern Python, dictionaries preserve insertion order.

## 1. `keys()`, `values()`, and `items()`

Dictionaries provide three important methods for accessing their contents:

- `keys()` returns a view of the dictionary's keys.
- `values()` returns a view of the dictionary's values.
- `items()` returns a view containing key-value pairs as tuples.

These views are iterable and are commonly used in loops.

````python
student = {
    "name": "Alice",
    "age": 21,
    "score": 92
}

# Get all keys
print(student.keys())

# Get all values
print(student.values())

# Get key-value pairs
print(student.items())

# Iterate over keys
for key in student.keys():
    print(key)

# Iterate over values
for value in student.values():
    print(value)

# Iterate over both key and value
for key, value in student.items():
    print(f"{key}: {value}")
````

### Illustrative Example

A dictionary can represent configuration values for a program.

````python
config = {
    "learning_rate": 0.001,
    "batch_size": 32,
    "epochs": 10
}

for parameter, value in config.items():
    print(f"{parameter} = {value}")
````

### Edge Cases

`keys()`, `values()`, and `items()` return view objects rather than ordinary lists.

If a list is specifically required, convert the view:

```python
keys = list(student.keys())
values = list(student.values())
items = list(student.items())
```

The views are dynamic, so they can reflect changes made to the dictionary.

## 2. `get()` vs Direct Key Access

There are two common ways to retrieve a value from a dictionary.

### Direct key access

```python
value = data["key"]
```

If the key does not exist, Python raises `KeyError`.

### `get()`

```python
value = data.get("key")
```

If the key does not exist, `get()` returns `None` by default.

A default value can also be supplied:

```python
value = data.get("key", default_value)
```

````python
user = {
    "name": "Alice",
    "age": 21
}

# Direct access
print(user["name"])

# get() access
print(user.get("name"))

# Missing key with get()
print(user.get("email"))

# Missing key with a custom default
print(user.get("email", "Not provided"))

# Direct access to a missing key would raise KeyError
# print(user["email"])
````

### Illustrative Example

`get()` is useful when working with optional fields.

````python
model_config = {
    "learning_rate": 0.001,
    "batch_size": 32
}

learning_rate = model_config.get("learning_rate", 0.01)
optimizer = model_config.get("optimizer", "adam")

print("Learning rate:", learning_rate)
print("Optimizer:", optimizer)
````

### Edge Cases

`get()` does not distinguish between a missing key and a key explicitly mapped to `None` unless you check membership separately.

```python
data = {"result": None}

print(data.get("result"))  # None
print(data.get("missing")) # None
```

To distinguish them:

```python
if "result" in data:
    print("The key exists.")
```

## 3. `update()`

`update()` adds key-value pairs to a dictionary or replaces values for keys that already exist.

It can accept another dictionary or an iterable of key-value pairs.

```python
data.update({"a": 10, "b": 20})
```

If a key already exists, its value is replaced.

````python
user = {
    "name": "Alice",
    "age": 21
}

# Add a new key
user.update({"city": "Hyderabad"})

# Replace an existing value
user.update({"age": 22})

print(user)

# update() can also accept keyword arguments
user.update(country="India", active=True)

print(user)

# update() accepts an iterable of pairs
user.update([
    ("role", "student"),
    ("level", 3)
])

print(user)
````

### Illustrative Example

`update()` is useful for applying configuration overrides.

````python
default_config = {
    "learning_rate": 0.001,
    "batch_size": 32,
    "epochs": 10
}

user_config = {
    "batch_size": 64,
    "epochs": 20
}

default_config.update(user_config)

print(default_config)
````

### Edge Cases

`update()` mutates the existing dictionary and returns `None`.

```python
result = user.update({"age": 23})

print(result)  # None
```

Do not expect `update()` to return the updated dictionary.

## 4. `pop()` and `popitem()`

### `pop()`

`pop(key)` removes a key and returns its associated value.

```python
value = data.pop("key")
```

A default value can be supplied for a missing key:

```python
value = data.pop("missing", default)
```

### `popitem()`

`popitem()` removes and returns the last inserted key-value pair.

It returns a tuple:

```python
(key, value)
```

Calling `popitem()` on an empty dictionary raises `KeyError`.

````python
data = {
    "name": "Alice",
    "age": 21,
    "city": "Hyderabad"
}

# Remove a specific key and return its value
age = data.pop("age")

print("Removed age:", age)
print("Dictionary:", data)

# Remove the last inserted item
item = data.popitem()

print("Removed item:", item)
print("Dictionary:", data)

# Missing key with a default
result = data.pop("email", "Not found")
print(result)
````

### Illustrative Example

`pop()` can be useful when processing data where a field needs to be extracted and removed.

````python
record = {
    "id": 101,
    "name": "Alice",
    "score": 95
}

score = record.pop("score")

print("Score:", score)
print("Remaining record:", record)
````

### Edge Cases

Without a default, `pop()` raises `KeyError` for a missing key:

```python
data = {"a": 1}

# data.pop("b")  # KeyError
```

`popitem()` removes the most recently inserted item in modern Python.

```python
data = {"a": 1, "b": 2}

print(data.popitem())  # ('b', 2)
```

## 5. `setdefault()`

`setdefault()` retrieves a value if a key exists.

If the key does not exist, it inserts the key with the supplied default value and returns that default.

Syntax:

```python
dictionary.setdefault(key, default)
```

This is particularly useful for building dictionaries containing lists or other mutable collections.

````python
data = {
    "name": "Alice"
}

# Existing key: returns its existing value
name = data.setdefault("name", "Unknown")
print(name)
print(data)

# Missing key: inserts the default value
city = data.setdefault("city", "Hyderabad")
print(city)
print(data)
````

### Illustrative Example

A common use is grouping values into lists.

````python
groups = {}

groups.setdefault("even", []).append(2)
groups.setdefault("even", []).append(4)
groups.setdefault("odd", []).append(1)
groups.setdefault("odd", []).append(3)

print(groups)
````

### Edge Cases

Be careful when using mutable defaults.

`setdefault()` inserts the exact object supplied as the default when the key is missing.

For simple grouping code, the following is usually fine:

```python
groups.setdefault("items", []).append(value)
```

For more complex dictionary construction, `collections.defaultdict` can sometimes provide clearer code.

## 6. Dictionary Comprehension

Dictionary comprehensions provide a compact way to create dictionaries.

Basic syntax:

```python
{key_expression: value_expression for item in iterable}
```

A condition can be added:

```python
{key: value for item in iterable if condition}
```

````python
numbers = [1, 2, 3, 4, 5]

# Map each number to its square
squares = {
    number: number ** 2
    for number in numbers
}

print(squares)

# Include only even numbers
even_squares = {
    number: number ** 2
    for number in numbers
    if number % 2 == 0
}

print(even_squares)
````

### Illustrative Example

Dictionary comprehensions are useful for transforming labeled data.

````python
names = ["alice", "bob", "charlie"]

name_lengths = {
    name: len(name)
    for name in names
}

print(name_lengths)
````

### Edge Cases

If a comprehension generates the same key more than once, the later value overwrites the earlier value.

```python
values = [1, 2, 3]

result = {x % 2: x for x in values}

print(result)
# {1: 3, 0: 2}
```

For complicated transformations, a normal loop may be more readable.

## 7. Iterating Through Key-Value Pairs

The most common way to iterate over both dictionary keys and values is:

```python
for key, value in dictionary.items():
    ...
```

This is usually clearer than manually looking up each value using the key.

````python
student_scores = {
    "Alice": 92,
    "Bob": 85,
    "Charlie": 96
}

for name, score in student_scores.items():
    print(f"{name}: {score}")
````

### Illustrative Example

Filtering dictionary entries while iterating can be done with a condition.

````python
scores = {
    "Alice": 92,
    "Bob": 68,
    "Charlie": 88,
    "Diana": 74
}

for name, score in scores.items():
    if score >= 80:
        print(f"{name} passed with {score}")
````

### Edge Cases

Avoid changing the dictionary's size while directly iterating over it.

This can raise `RuntimeError`:

```python
data = {"a": 1, "b": 2}

# for key in data:
#     del data[key]
```

A safer approach is to iterate over a snapshot:

```python
for key in list(data):
    del data[key]
```

Alternatively, build a new dictionary using a comprehension.

## 8. Dictionary Merging

Python provides several ways to combine dictionaries.

### `update()`

Mutates an existing dictionary:

```python
a.update(b)
```

### Dictionary unpacking

Creates a new dictionary:

```python
merged = {**a, **b}
```

If the same key exists in both dictionaries, the later value wins.

### Union operator `|`

Python 3.9+ supports:

```python
merged = a | b
```

This creates a new dictionary.

### In-place union `|=`

Python 3.9+ also supports:

```python
a |= b
```

This updates `a` in place.

````python
first = {
    "name": "Alice",
    "age": 21
}

second = {
    "age": 22,
    "city": "Hyderabad"
}

# update() modifies the first dictionary
updated = first.copy()
updated.update(second)

print("Using update():", updated)

# Dictionary unpacking creates a new dictionary
merged_unpacking = {**first, **second}

print("Using unpacking:", merged_unpacking)

# Union operator creates a new dictionary
merged_union = first | second

print("Using |:", merged_union)

# In-place union
first_copy = first.copy()
first_copy |= second

print("Using |=:", first_copy)
````

### Illustrative Example

Merging a default configuration with user-specific settings is a common pattern.

````python
default_settings = {
    "theme": "dark",
    "language": "en",
    "notifications": True
}

user_settings = {
    "language": "fr",
    "notifications": False
}

settings = default_settings | user_settings

print(settings)
````

### Edge Cases

When keys overlap, the dictionary on the right takes precedence:

```python
a = {"x": 1}
b = {"x": 2}

print(a | b)
# {'x': 2}
```

Dictionary union requires Python 3.9 or later.

## 9. Nested Dictionary Modification

Dictionary values can themselves be dictionaries, allowing hierarchical data structures.

Example:

```python
student = {
    "name": "Alice",
    "academic": {
        "math": 90,
        "science": 95
    }
}
```

Nested values can be accessed and modified by chaining keys.

````python
student = {
    "name": "Alice",
    "academic": {
        "math": 90,
        "science": 95
    }
}

# Read a nested value
print(student["academic"]["math"])

# Modify a nested value
student["academic"]["math"] = 94

print(student)

# Add a new nested key
student["academic"]["english"] = 88

print(student)
````

### Illustrative Example

Nested dictionaries are useful for representing structured records.

````python
model = {
    "name": "Neural Network",
    "architecture": {
        "input_size": 10,
        "hidden_layers": [64, 32],
        "output_size": 3
    },
    "training": {
        "epochs": 20,
        "batch_size": 32
    }
}

# Change a nested training parameter
model["training"]["batch_size"] = 64

# Add a new training parameter
model["training"]["learning_rate"] = 0.001

print(model)
````

### Edge Cases

Chained indexing raises `KeyError` if an intermediate key is missing.

```python
# student["missing"]["score"]  # KeyError
```

If nested data may be incomplete, check keys or use methods such as `get()` carefully.

Also, if an intermediate value is `None` or has an unexpected type, chained access can raise `TypeError`.

## 10. Dictionary Copying and References

Like lists, dictionaries are mutable objects.

Assignment creates another reference to the same dictionary:

```python
a = {"x": 1}
b = a
```

A shallow copy creates a new outer dictionary:

```python
b = a.copy()
```

For nested dictionaries, the nested mutable objects remain shared in a shallow copy.

`copy.deepcopy()` recursively copies the structure.

````python
from copy import deepcopy

original = {
    "name": "Alice",
    "scores": {
        "math": 90,
        "science": 95
    }
}

# Assignment creates an alias
alias = original
alias["name"] = "Bob"

print("Original after alias modification:", original)

# Restore original for the next examples
original = {
    "name": "Alice",
    "scores": {
        "math": 90,
        "science": 95
    }
}

# Shallow copy
shallow = original.copy()
shallow["name"] = "Charlie"

print("Original after shallow outer modification:", original)
print("Shallow copy:", shallow)

# Nested dictionary is still shared
shallow["scores"]["math"] = 99

print("Original after nested modification:", original)
print("Shallow copy:", shallow)

# Deep copy
original = {
    "name": "Alice",
    "scores": {
        "math": 90,
        "science": 95
    }
}

deep = deepcopy(original)
deep["scores"]["math"] = 100

print("Original after deep modification:", original)
print("Deep copy:", deep)
````

### Illustrative Example

Deep copying is useful when creating an independent configuration from a nested template.

````python
from copy import deepcopy

base_config = {
    "model": {
        "layers": [128, 64],
        "dropout": 0.2
    },
    "training": {
        "batch_size": 32,
        "epochs": 10
    }
}

experiment_config = deepcopy(base_config)

experiment_config["model"]["layers"].append(32)
experiment_config["training"]["batch_size"] = 64

print("Base configuration:")
print(base_config)

print("\nExperiment configuration:")
print(experiment_config)
````

### Edge Cases

The distinction between assignment, shallow copy, and deep copy is especially important for nested dictionaries.

```python
a = {"nested": {"value": 10}}

b = a
c = a.copy()
d = deepcopy(a)
```

- `b` shares the entire dictionary with `a`.
- `c` has a new outer dictionary but shares the nested dictionary.
- `d` recursively copies the nested structure.

Deep copying can require more memory and processing time than a shallow copy.

# Combined Example

The following example combines several dictionary concepts in a small data-processing workflow.

````python
students = [
    {"name": "Alice", "score": 92, "department": "CSE"},
    {"name": "Bob", "score": 78, "department": "ECE"},
    {"name": "Charlie", "score": 95, "department": "CSE"},
]

# Create a dictionary mapping names to scores
score_lookup = {
    student["name"]: student["score"]
    for student in students
}

print("Score lookup:", score_lookup)

# Update a student's score
score_lookup.update({"Bob": 82})

# Safely retrieve a score
print("Bob's score:", score_lookup.get("Bob", 0))

# Iterate through key-value pairs
for name, score in score_lookup.items():
    print(f"{name}: {score}")

# Filter high scorers
high_scorers = {
    name: score
    for name, score in score_lookup.items()
    if score >= 85
}

print("High scorers:", high_scorers)

# Merge additional information
extra_data = {
    "course": "Python",
    "semester": 3
}

metadata = {
    "source": "notebook"
} | extra_data

print("Metadata:", metadata)
````

# Quick Reference

| Operation | Example | Purpose |
|---|---|---|
| `keys()` | `d.keys()` | Get dictionary keys |
| `values()` | `d.values()` | Get dictionary values |
| `items()` | `d.items()` | Get key-value pairs |
| Direct access | `d["key"]` | Access a required key |
| `get()` | `d.get("key")` | Safely access a possibly missing key |
| `update()` | `d.update(other)` | Add or replace entries |
| `pop()` | `d.pop("key")` | Remove a specific key |
| `popitem()` | `d.popitem()` | Remove the last inserted pair |
| `setdefault()` | `d.setdefault("key", [])` | Get or initialize a key |
| Comprehension | `{k: v for ...}` | Build a dictionary |

| Nested access | `d["a"]["b"]` | Access nested data |
| Shallow copy | `d.copy()` | Copy outer dictionary |
| Deep copy | `deepcopy(d)` | Recursively copy nested objects |

# Common Dictionary Pitfalls

### 1. Missing keys

```python
data = {"name": "Alice"}

# data["age"]  # KeyError
```

Use `get()` when the key is optional.

### 2. Forgetting that dictionaries are mutable

Methods such as `update()`, `pop()`, and `setdefault()` modify the dictionary.

### 3. Confusing a view with a list

```python
keys = data.keys()
```

does not create an ordinary list.

### 4. Modifying dictionary size during iteration

Avoid adding or removing keys directly while iterating over the same dictionary.

### 5. Shallow copies of nested dictionaries

`.copy()` only copies the outer dictionary.

### 6. Duplicate keys

Dictionary keys must be unique. If duplicate keys occur when constructing a dictionary, the later value replaces the earlier one:

```python
data = {
    "x": 1,
    "x": 2
}

print(data)
# {'x': 2}
```

### 7. Dictionary keys must be hashable

Immutable/hashable objects such as strings, numbers, and tuples containing hashable values can be keys.

Mutable objects such as lists cannot be keys:

```python
# data = {[1, 2]: "value"}  # TypeError
```
