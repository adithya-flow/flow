## 2. `__init__` and `self`

### Explanation
`__init__` initializes an instance after it has been created. `self` refers to the current instance.

### Edge cases
`self` is a convention, not a keyword. `__init__` is an initializer; `__new__` is involved in creating the instance.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

s = Student("Asha", 20)
print(s.name, s.age)
```

### Illustrative example
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

a = BankAccount("Ravi", 5000)
a.balance += 1000
print(a.owner, a.balance)

## 3. Instance Attributes vs Class Attributes

### Explanation
Instance attributes belong to individual objects. Class attributes are defined on the class and are shared through the class unless an instance shadows them.

### Edge cases
Mutable class attributes such as `items = []` are shared. Assignment through an instance can create an instance attribute that shadows a class attribute.

```python
class Employee:
    company = "Example Corp"

    def __init__(self, name):
        self.name = name

a = Employee("Asha")
b = Employee("Ravi")
a.company = "Other Corp"
print(a.company, b.company, Employee.company)
```

### Illustrative example

        self.name = name

## 4. Instance Methods, `@classmethod`, and `@staticmethod`

### Explanation
Instance methods receive `self`. Class methods receive `cls` and are useful for class-level behavior and alternate constructors. Static methods receive neither automatically and are useful for utilities grouped with a class.

### Edge cases
A class method receives the subclass through `cls` when invoked through a subclass. A static method has no automatic instance/class context.

```python
class Employee:
    company = "Example Corp"

    def __init__(self, name):
        self.name = name

    def describe(self):
        return f"{self.name} works at {self.company}"

    @classmethod
    def company_name(cls):
        return cls.company

    @staticmethod
    def valid_name(name):
        return isinstance(name, str) and bool(name.strip())

e = Employee("Asha")
print(e.describe())
print(Employee.company_name())
print(Employee.valid_name("Ravi"))
```

### Illustrative example
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    @classmethod
    def from_string(cls, text):
        x, y = map(float, text.split(","))
        return cls(x, y)

p = Point.from_string("3,4")
print(p.x, p.y)

## 5. Encapsulation: public, `_protected`, and `__private`

### Explanation
Python uses conventions rather than enforced access modifiers. `name` is public, `_name` signals non-public use, and `__name` triggers name mangling.

### Edge cases
`_name` is not protected by the runtime. `__name` is not true security; it becomes a mangled attribute such as `_Account__pin`.

```python
class Account:
    def __init__(self):
        self.owner = "Asha"
        self._balance = 1000
        self.__pin = 1234

a = Account()
print(a.owner)
print(a._balance)
print(a._Account__pin)  # name-mangled access
```

### Illustrative example
class User:
    def __init__(self, username, password):
        self.username = username
        self.__password = password

    def check_password(self, value):
        return value == self.__password

print(User("alice", "secret").check_password("secret"))

## 6. `@property`: getters and setters

### Explanation
Properties provide attribute-style access while allowing validation or computed behavior.

### Edge cases
A property without a setter is normally read-only through normal assignment. Setters are useful for enforcing invariants.

```python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("Below absolute zero")
        self._celsius = value

    @property
    def fahrenheit(self):
        return self.celsius * 9 / 5 + 32

t = Temperature(25)
print(t.celsius, t.fahrenheit)
```

### Illustrative example
class Rectangle:
    def __init__(self, width, height):
        self.width, self.height = width, height

    @property
    def area(self):
        return self.width * self.height

print(Rectangle(5, 4).area)

## 7. Inheritance: single, multilevel, and multiple

### Explanation
Inheritance represents an `is-a` relationship. Single inheritance has one direct base, multilevel forms a chain, and multiple inheritance has multiple bases.

### Edge cases
Multiple inheritance requires attention to MRO. Deep inheritance trees can become difficult to maintain; composition is often simpler.

```python
class Animal:
    def speak(self):
        return "sound"

class Dog(Animal):
    pass

class Puppy(Dog):
    def play(self):
        return "playing"

class Swimmer:
    def swim(self): return "swimming"
class Flyer:
    def fly(self): return "flying"
class Duck(Swimmer, Flyer):
    pass

print(Dog().speak())
print(Puppy().play())
print(Duck().swim(), Duck().fly())
```

### Illustrative example
class Model:
    def predict(self, x):
        raise NotImplementedError

class Classifier(Model):
    def predict(self, x):
        return "class-A"

print(Classifier().predict("image"))

## 8. Method Overriding and `super()`

### Explanation
A subclass can override an inherited method. `super()` accesses the next implementation in the MRO and is commonly used to extend parent initialization or behavior.

### Edge cases
`super()` means the next implementation according to the MRO, not simply a hard-coded parent. An override does not automatically call the parent.

```python
class Animal:
    def __init__(self, name):
        self.name = name
    def speak(self):
        return "sound"

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed
    def speak(self):
        return "woof"

d = Dog("Max", "Labrador")
print(d.name, d.breed, d.speak())
```

### Illustrative example
class Logger:
    def save(self, message):
        print("Saving:", message)

class AuditedLogger(Logger):
    def save(self, message):
        print("Audit entry")
        super().save(message)

AuditedLogger().save("model trained")

## 9. Method Resolution Order (MRO)

### Explanation
MRO defines the order Python uses to search classes for methods and attributes. Python uses C3 linearization for multiple inheritance.

### Edge cases
In cooperative multiple inheritance, methods should consistently call `super()`. An inconsistent inheritance graph can raise `TypeError` because no valid MRO exists.

```python
class A:
    def show(self): return "A"
class B(A):
    def show(self): return "B"
class C(A):
    def show(self): return "C"
class D(B, C):
    pass

print(D.__mro__)
print(D().show())
```

### Illustrative example
class Base:
    def action(self): print("Base")
class Left(Base):
    def action(self):
        print("Left"); super().action()
class Right(Base):
    def action(self):
        print("Right"); super().action()
class Child(Left, Right):
    def action(self):
        print("Child"); super().action()

Child().action()

## 10. Polymorphism and Duck Typing

### Explanation
Polymorphism lets different objects be used through a common interface. Duck typing focuses on supported behavior rather than requiring a specific inheritance hierarchy.

### Edge cases
Duck typing can fail with `AttributeError` when an expected operation is absent. Explicit checks are useful at boundaries, but often unnecessary for ordinary polymorphic code.

```python
class Dog:
    def speak(self): return "woof"
class Cat:
    def speak(self): return "meow"

def make_sound(animal):
    return animal.speak()

print(make_sound(Dog()))
print(make_sound(Cat()))
```

### Illustrative example
class CSVExporter:
    def export(self, data): return "csv data"
class JSONExporter:
    def export(self, data): return "json data"

def save_report(exporter, data):
    return exporter.export(data)

print(save_report(CSVExporter(), [1, 2]))
print(save_report(JSONExporter(), [1, 2]))

## 11. Abstraction with `abc`

### Explanation
The `abc` module supports explicit abstract interfaces. `ABC` provides a base class and `@abstractmethod` marks required operations.

### Edge cases
An incomplete abstract class cannot normally be instantiated. Abstract methods may still have implementations that subclasses can call with `super()`.

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self, w, h):
        self.w, self.h = w, h
    def area(self):
        return self.w * self.h

print(Rectangle(5, 4).area())
```

### Illustrative example
from abc import ABC, abstractmethod

class Model(ABC):
    @abstractmethod
    def predict(self, x): ...

class LinearModel(Model):
    def predict(self, x): return 2 * x + 1

print(LinearModel().predict(5))

## 12. Composition vs Inheritance: “has-a” vs “is-a”

### Explanation
Inheritance models `is-a`; composition models `has-a`. Composition lets components be replaced independently.

### Edge cases
Use inheritance when the subtype genuinely conforms to the parent abstraction. Composition is often better when behavior is replaceable or independent.

```python
class Engine:
    def start(self): return "engine started"

class Car:
    def __init__(self, engine): self.engine = engine
    def start(self): return self.engine.start()

print(Car(Engine()).start())
```

### Illustrative example
class Tokenizer:
    def tokenize(self, text): return text.lower().split()

class TextClassifier:
    def __init__(self, tokenizer): self.tokenizer = tokenizer
    def predict(self, text):
        return "positive" if "good" in self.tokenizer.tokenize(text) else "unknown"

print(TextClassifier(Tokenizer()).predict("This is good"))

## 13. Dunder (Magic) Methods

### Explanation
Dunder methods define Python data-model behavior. Common examples include `__str__`, `__repr__`, `__len__`, `__eq__`, `__add__`, `__getitem__`, and `__call__`.

### Edge cases
Returning `NotImplemented` from binary methods is different from raising `NotImplementedError`: it lets Python try other supported operation paths. `__repr__` should be useful for debugging.

```python
class Vector:
    def __init__(self, x, y): self.x, self.y = x, y
    def __str__(self): return f"Vector({self.x}, {self.y})"
    def __repr__(self): return f"Vector(x={self.x!r}, y={self.y!r})"
    def __len__(self): return 2
    def __eq__(self, other):
        if not isinstance(other, Vector): return NotImplemented
        return (self.x, self.y) == (other.x, other.y)
    def __add__(self, other):
        if not isinstance(other, Vector): return NotImplemented
        return Vector(self.x + other.x, self.y + other.y)

v = Vector(1, 2)
print(str(v), repr(v), len(v), v == Vector(1, 2), v + Vector(3, 4))
```

### Illustrative example
class Dataset:
    def __init__(self, values): self.values = list(values)
    def __getitem__(self, index): return self.values[index]
    def __len__(self): return len(self.values)

class Multiplier:
    def __init__(self, factor): self.factor = factor
    def __call__(self, value): return value * self.factor

d = Dataset([10, 20, 30])
print(d[1], len(d), Multiplier(3)(7))

## 14. `isinstance()` and `issubclass()`

### Explanation
`isinstance(obj, Class)` checks an object against a class and its subclasses. `issubclass(A, B)` checks whether class `A` derives from `B`.

### Edge cases
A subclass instance is also an instance of its parent. `isinstance(x, (A, B))` accepts multiple candidate classes. `bool` is a subclass of `int`.

```python
class Animal: pass
class Dog(Animal): pass

d = Dog()
print(isinstance(d, Dog))
print(isinstance(d, Animal))
print(issubclass(Dog, Animal))
```

### Illustrative example
class Model: pass
class Classifier(Model): pass

m = Classifier()
print(isinstance(m, Model))
print(issubclass(Classifier, Model))

## 15. `@dataclass`

### Explanation
`@dataclass` generates common boilerplate from declared fields, including `__init__`, `__repr__`, and `__eq__` by default.

### Edge cases
Use `default_factory` for mutable defaults. `frozen=True` prevents normal field assignment. Dataclasses are best suited to data-centric classes.

```python
from dataclasses import dataclass

@dataclass
class Point:
    x: float
    y: float

p1 = Point(1, 2)
p2 = Point(1, 2)
print(p1)
print(p1 == p2)
```

### Illustrative example
from dataclasses import dataclass, field

@dataclass
class TrainingConfig:
    learning_rate: float
    batch_size: int = 32
    tags: list[str] = field(default_factory=list)

c = TrainingConfig(0.001)
c.tags.append("baseline")
print(c)

## 16. Optional Advanced: `__slots__`

### Explanation
`__slots__` declares allowed instance attributes and can remove the normal instance `__dict__`, reducing memory overhead and preventing accidental attributes.

### Edge cases
Slotted classes normally have no `__dict__` unless requested. Multiple inheritance has additional slot-layout constraints. `__slots__` is an optimization/design choice, not security.

```python
class Point:
    __slots__ = ("x", "y")
    def __init__(self, x, y): self.x, self.y = x, y

p = Point(1, 2)
print(p.x, p.y)
try:
    p.z = 3
except AttributeError as e:
    print(type(e).__name__, e)
```

### Illustrative example
class Record:
    __slots__ = ("name", "value")
    def __init__(self, name, value): self.name, self.value = name, value

print(Record("accuracy", 0.95).value)

## 17. Optional Advanced: `__new__`

### Explanation
`__new__` creates and returns an instance; `__init__` then initializes it. `__new__` is especially relevant for immutable built-in subclasses and custom instance creation.

### Edge cases
If `__new__` returns an instance of the class, Python normally calls `__init__`. Immutable types such as `int`, `str`, and `tuple` commonly require customization in `__new__`.

```python
class Demo:
    def __new__(cls, *args, **kwargs):
        print("__new__")
        return super().__new__(cls)
    def __init__(self):
        print("__init__")

Demo()
```

### Illustrative example
class PositiveInt(int):
    def __new__(cls, value):
        value = int(value)
        if value < 0: raise ValueError("value must be non-negative")
        return super().__new__(cls, value)

print(PositiveInt(10))

## 18. Optional Advanced: Operator Overloading in Depth

### Explanation
Operators map to special methods: `+` → `__add__`, `-` → `__sub__`, `*` → `__mul__`, comparisons → methods such as `__eq__` and `__lt__`, containment → `__contains__`, and indexing → `__getitem__`.

### Edge cases
Return `NotImplemented` for unsupported operand types. Operator overloads should have intuitive semantics. Defining `__eq__` also affects hashing behavior and whether instances can be used as dictionary keys.

```python
class Money:
    def __init__(self, amount): self.amount = amount
    def __add__(self, other):
        if isinstance(other, Money): return Money(self.amount + other.amount)
        if isinstance(other, (int, float)): return Money(self.amount + other)
        return NotImplemented
    def __radd__(self, other): return self + other
    def __repr__(self): return f"Money({self.amount})"

a = Money(100)
print(a + Money(50))
print(a + 25)
print(25 + a)
```

### Illustrative example
class Score:
    def __init__(self, value): self.value = value
    def __eq__(self, other):
        if not isinstance(other, Score): return NotImplemented
        return self.value == other.value
    def __lt__(self, other):
        if not isinstance(other, Score): return NotImplemented
        return self.value < other.value

scores = [Score(70), Score(95), Score(82)]
scores.sort()
print([s.value for s in scores])

## 19. Practical OOP Example: Model Registry

This combines abstraction, class methods, properties, inheritance, polymorphism, and composition.

```python
from abc import ABC, abstractmethod

class Preprocessor:
    def transform(self, text):
        return text.lower().strip()

class Model(ABC):
    registry = {}

    def __init__(self, name):
        self.name = name
        self._trained = False

    @property
    def trained(self):
        return self._trained

    @classmethod
    def register(cls, model_cls):
        cls.registry[model_cls.__name__] = model_cls
        return model_cls

    @abstractmethod
    def predict(self, text):
        pass

@Model.register
class KeywordModel(Model):
    def __init__(self, name, preprocessor=None):
        super().__init__(name)
        self.preprocessor = preprocessor or Preprocessor()

    def predict(self, text):
        text = self.preprocessor.transform(text)
        return "positive" if "good" in text else "unknown"

model = KeywordModel("demo")
print(model.predict("This is GOOD"))
print(Model.registry)
print(model.trained)
```

## 20. Common Pitfalls

- **Mutable class attributes:** `items = []` is shared by instances; use an instance attribute for independent state.
- **Shadowing class attributes:** `obj.attr = value` can create an instance attribute that hides the class attribute.
- **Missing `super().__init__()`:** a subclass initializer may need to initialize required parent state.
- **Overusing inheritance:** use composition when the relationship is not truly `is-a`.
- **Misunderstanding `__private`:** name mangling is not security.
- **Unexpected operator overloads:** magic methods should follow intuitive semantics.
- **Confusing `__new__` and `__init__`:** creation and initialization are separate stages.
- **Property syntax:** a property is accessed as `obj.value`, not `obj.value()`.
