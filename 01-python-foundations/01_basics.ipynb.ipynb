## 1. Variables and Naming Conventions

Variables are names bound to Python objects. Python does not require a variable declaration or explicit type annotation before assignment.

Common naming conventions:
- `snake_case` for variables and functions
- `UPPER_CASE` for constants by convention
- `PascalCase` for classes
- Avoid shadowing built-ins such as `list`, `str`, `sum`, or `id`

````python
# Variables are created by assignment
name = "Alice"
age = 21
height_m = 1.68

print(name, age, height_m)

# Multiple assignment
x, y = 10, 20

# Swapping values without a temporary variable
x, y = y, x
print(x, y)

# Constants are conventions, not enforced by Python
PI = 3.14159
````

### Practical example

Use descriptive variable names when storing values that represent different parts of a program.

````python
student_name = "Aarav"
course_name = "Python"
completed_lessons = 12

print(f"{student_name} completed {completed_lessons} lessons in {course_name}.")
````

### Edge cases and common mistakes

- A variable name cannot start with a digit.
- Names are case-sensitive: `age`, `Age`, and `AGE` are different.
- Do not overwrite important built-ins accidentally, such as `list = []`.
- Assignment binds a name to an object; it does not copy the object automatically.

## 2. Built-in Data Types

Python provides several fundamental built-in types:

| Type | Example | Description |
|---|---|---|
| `int` | `42` | Integer |
| `float` | `3.14` | Floating-point number |
| `complex` | `2 + 3j` | Complex number |
| `bool` | `True` | Boolean value |
| `str` | `"hello"` | Text |
| `NoneType` | `None` | Absence of a value |

`bool` is a subclass of `int`, so `True` behaves numerically like `1` and `False` like `0` in arithmetic contexts.

````python
integer_value = 42
float_value = 3.14
complex_value = 2 + 3j
boolean_value = True
text_value = "hello"
nothing = None

print(integer_value)
print(float_value)
print(complex_value)
print(boolean_value)
print(text_value)
print(nothing)

# bool is a subclass of int
print(True + True)   # 2
print(False + 10)    # 10
````

### Practical example

````python
model_version = 3
accuracy = 0.947
learning_enabled = True
model_name = "classifier"
last_error = None

print(model_name, model_version, accuracy, learning_enabled, last_error)
````

### Edge cases

- `float` values are subject to floating-point representation limitations.
- Complex numbers use `j` for the imaginary component.
- `None` is a singleton object commonly used to represent “no value”.
- `bool` values are `True` and `False`, with capitalized names.

## 3. Type Checking with `type()`

`type()` returns the type object of a value or object.

For most everyday checks, `isinstance()` is preferred when inheritance should be considered, but `type()` is useful for inspecting the exact runtime type.

````python
value = 42

print(type(value))
print(type(value).__name__)

# Exact type comparison
print(type(value) is int)
print(type(True) is bool)
````

### Edge cases

`type(True) is int` is `False` even though `bool` is a subclass of `int`. For inheritance-aware checks, use `isinstance()`.

````python
print(type(True) is int)
print(isinstance(True, int))

class Parent:
    pass

class Child(Parent):
    pass

obj = Child()

print(type(obj) is Parent)       # False
print(isinstance(obj, Parent))   # True
````

## 4. Type Conversion

Common conversion functions include:
- `int()`
- `float()`
- `str()`
- `bool()`

Conversion may fail when the input cannot be interpreted as the requested type.

````python
print(int("42"))
print(float("3.14"))
print(str(100))
print(bool(1))

# Numeric conversions
print(int(3.99))      # Truncates toward zero
print(float(10))
print(bool(0))
print(bool(5))
````

### Practical example

````python
user_age = "21"

# Convert text input into a number before arithmetic
age = int(user_age)
next_year_age = age + 1

print(next_year_age)
````

### Edge cases

- `int(3.99)` produces `3`; it does not round.
- `int("3.14")` raises `ValueError`.
- `bool("")`, `bool(0)`, and `bool(None)` are `False`.
- `bool("False")` is `True` because the string is non-empty.

````python
print(bool(""))
print(bool("False"))

# This would raise ValueError:
# print(int("3.14"))
````

## 5. Arithmetic Operators

Python's main arithmetic operators are:

| Operator | Meaning |
|---|---|
| `+` | Addition |
| `-` | Subtraction |
| `*` | Multiplication |
| `/` | True division |
| `//` | Floor division |
| `%` | Modulo / remainder |
| `**` | Exponentiation |

````python
a = 17
b = 5

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** 2)
````

### Practical example

````python
total_items = 23
items_per_batch = 5

full_batches = total_items // items_per_batch
remaining_items = total_items % items_per_batch

print("Full batches:", full_batches)
print("Remaining items:", remaining_items)
````

### Edge cases

- `/` always produces a floating-point result for ordinary numeric operands.
- `//` performs floor division, not simple truncation for negative values.
- Division or modulo by zero raises `ZeroDivisionError`.
- Exponentiation can produce very large integers.

````python
print(-7 // 3)   # -3: floor toward negative infinity
print(-7 % 3)    # 2

# print(10 / 0)   # ZeroDivisionError
````

## 6. Comparison Operators

Comparison operators produce Boolean results:

`==`, `!=`, `<`, `<=`, `>`, `>=`

Use `==` for equality and `is` for object identity. They are not interchangeable.

````python
a = 10
b = 20

print(a == b)
print(a != b)
print(a < b)
print(a <= b)
print(a > b)
print(a >= b)
````

### Practical example

````python
score = 87
passing_score = 40

passed = score >= passing_score
print(passed)
````

### Edge cases

Comparisons can sometimes be chained. Python evaluates `a < b < c` as a combined comparison rather than as `(a < b) < c`.

````python
age = 21
print(18 <= age < 60)

# Equivalent logical form
print(age >= 18 and age < 60)
````

## 7. Logical Operators

Python provides:
- `and`
- `or`
- `not`

These operators work with truth values, but `and` and `or` return one of their operands rather than always returning `True` or `False`.

````python
x = True
y = False

print(x and y)
print(x or y)
print(not x)

# and/or return operands
print(0 or 42)
print("hello" and 123)
````

### Practical example

````python
username = "jaya"
is_authenticated = True

can_access = bool(username) and is_authenticated
print(can_access)
````

### Edge cases

Python short-circuits logical expressions:
- `A and B` does not evaluate `B` if `A` is falsy.
- `A or B` does not evaluate `B` if `A` is truthy.

This behavior is often used for conditional defaults and guarding expressions.

````python
def get_value():
    print("Function evaluated")
    return 100

print(False and get_value())  # get_value() is not called
print(True or get_value())    # get_value() is not called
````

## 8. Assignment Operators

Assignment operators combine assignment with an operation.

Common forms include:
`+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=`, `&=`, `|=`, `^=`, `<<=`, `>>=`.

````python
value = 10

value += 5
print(value)

value -= 2
print(value)

value *= 3
print(value)

value //= 4
print(value)

value %= 5
print(value)

value **= 2
print(value)
````

### Edge cases

Augmented assignment may mutate an object in place when the object's type supports it. This matters with mutable objects such as lists.

````python
numbers = [1, 2]
alias = numbers

numbers += [3, 4]

print(numbers)
print(alias)  # Same list was mutated
````

## 9. Bitwise Operators

Bitwise operators operate on integer bit patterns:

| Operator | Meaning |
|---|---|
| `&` | AND |
| `|` | OR |
| `^` | XOR |
| `~` | NOT / inversion |
| `<<` | Left shift |
| `>>` | Right shift |

````python
a = 0b1100
b = 0b1010

print(bin(a & b))
print(bin(a | b))
print(bin(a ^ b))
print(bin(~a))
print(bin(a << 1))
print(bin(a >> 1))
````

### Practical example

Bitwise operations are commonly used for flags, masks, packed values, and low-level algorithms.

````python
READ = 0b001
WRITE = 0b010
EXECUTE = 0b100

permissions = READ | WRITE

print(bin(permissions))
print(bool(permissions & READ))
print(bool(permissions & EXECUTE))
````

### Edge cases

`~x` uses Python's signed integer representation and is equivalent to `-x - 1`. It is therefore common to mask the result when working with a fixed-width bit representation.

````python
x = 5

print(~x)
print(~x == -x - 1)

# Example of masking to the lowest 4 bits
print(bin(~x & 0b1111))
````

## 10. Operator Precedence

Python evaluates operators according to precedence rules. Parentheses can be used to make the intended order explicit.

A simplified order from higher to lower precedence:

1. Parentheses / calls / indexing
2. Exponentiation `**`
3. Unary `+`, `-`, `~`
4. `*`, `/`, `//`, `%`
5. `+`, `-`
6. Shifts
7. Bitwise `&`
8. Bitwise `^`
9. Bitwise `|`
10. Comparisons
11. `not`
12. `and`
13. `or`

When readability matters, prefer parentheses instead of relying on subtle precedence.

````python
result_1 = 2 + 3 * 4
result_2 = (2 + 3) * 4
result_3 = 2 ** 3 * 4

print(result_1)
print(result_2)
print(result_3)
````

### Edge cases

Exponentiation has special interaction with unary operators. For example, `-2 ** 2` is interpreted as `-(2 ** 2)`, while `(-2) ** 2` is positive.

````python
print(-2 ** 2)
print((-2) ** 2)
````

## 11. Truthy and Falsy Values

Python uses truth-value testing in `if`, `while`, and logical expressions.

Common falsy values include:
- `False`
- `None`
- numeric zero values such as `0`, `0.0`, and `0j`
- empty strings
- empty lists, tuples, sets, and dictionaries

Most other objects are truthy by default.

````python
values = [
    False,
    None,
    0,
    0.0,
    "",
    [],
    [1],
    "hello",
    42,
]

for value in values:
    print(repr(value), "->", bool(value))
````

### Practical example

````python
items = []

if items:
    print("Items are available.")
else:
    print("The collection is empty.")
````

### Edge cases

Custom classes can define their truth value using `__bool__()` or `__len__()`. An object can therefore be falsy even when it is not one of the built-in empty containers.

````python
class EmptyLike:
    def __bool__(self):
        return False

obj = EmptyLike()
print(bool(obj))
````

## 13. String Formatting with f-Strings

F-strings provide a concise way to embed expressions inside string literals.

Syntax:
```python
f"Text {expression}"
```

They also support formatting specifications such as decimal precision, alignment, width, and percentage formatting.

````python
name = "Alice"
score = 0.93456

print(f"Name: {name}")
print(f"Score: {score:.2%}")
print(f"Rounded score: {score:.3f}")
````

### Practical example

````python
model_name = "classifier"
accuracy = 0.9472
samples = 12500

report = (
    f"Model: {model_name}\n"
    f"Accuracy: {accuracy:.2%}\n"
    f"Samples: {samples:,}"
)

print(report)
````

### Edge cases

Expressions inside `{}` are evaluated at runtime. Complex expressions are valid, but long expressions can reduce readability; calculate intermediate values when appropriate.

````python
items = [10, 20, 30]

print(f"Count: {len(items)}")
print(f"Total: {sum(items)}")
````

## 14. `help()`, `type()`, `dir()`, and `id()`

These built-in functions are useful for introspection:

- `help(obj)` — displays documentation
- `type(obj)` — returns the object's type
- `dir(obj)` — lists many available attributes and methods
- `id(obj)` — returns the object's identity value for its lifetime

`id()` is mainly useful for inspecting object identity; do not treat it as a portable memory address.

````python
text = "hello"

print(type(text))
print(id(text))

# Inspect available attributes/methods
print([name for name in dir(text) if name.startswith("upper")])

# Documentation can be inspected interactively:
# help(str.upper)
# help(list)
````

### Practical example

````python
numbers = [1, 2, 3]

print("Type:", type(numbers))
print("Identity:", id(numbers))
print("Has append:", "append" in dir(numbers))

# In Jupyter, help can be used directly:
# help(numbers.append)
````

### Edge cases

- `dir()` is intended for interactive exploration and does not necessarily represent every attribute available through dynamic behavior.
- `id()` values are only guaranteed to be unique among simultaneously existing objects.
- `help()` opens or displays documentation depending on the environment.

# Combined Example

The following example combines variables, type conversion, arithmetic, comparisons, truthiness, and f-string formatting in one small program.

````python
raw_price = "1499.50"
quantity = 3
discount_rate = 0.10

price = float(raw_price)
subtotal = price * quantity
discount = subtotal * discount_rate
final_price = subtotal - discount

is_valid_quantity = quantity > 0
has_discount = bool(discount_rate)

print(f"Subtotal: ₹{subtotal:,.2f}")
print(f"Discount: ₹{discount:,.2f}")
print(f"Final price: ₹{final_price:,.2f}")
print(f"Valid quantity: {is_valid_quantity}")
print(f"Discount applied: {has_discount}")
````

# Common Pitfalls

- `=` assigns a value; `==` compares values.
- `is` checks identity; `==` checks equality.
- `input()` returns a string.
- `/` and `//` are different: true division vs floor division.
- `bool("False")` is `True` because the string is non-empty.
- `int(3.9)` truncates toward zero; it does not round.
- `and` and `or` return operands and use short-circuit evaluation.
- `~x` is bitwise inversion, not simply “flip the displayed bits”.
- Assignment does not automatically copy mutable objects.
- Avoid shadowing built-in names such as `list`, `str`, `sum`, `id`, and `type`.
