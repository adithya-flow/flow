# Lists

Lists are ordered, mutable collections of objects.

They are one of Python's most commonly used data structures and are written using square brackets:

```python
numbers = [10, 20, 30]
```

A list can contain values of different types, including other lists:

```python
items = [10, "Python", 3.14, True]
nested = [[1, 2], [3, 4]]
```

This notebook covers indexing, slicing, comprehensions, nested lists, copying, useful list methods, iteration helpers, unpacking, and aliasing.

## 1. Positive and Negative Indexing

List indexing is zero-based.

- Positive indexing starts from the beginning: `0, 1, 2, ...`
- Negative indexing starts from the end: `-1, -2, -3, ...`

For a list with `n` elements, valid positive indexes are `0` through `n - 1`, while valid negative indexes are `-1` through `-n`.

````python
fruits = ["apple", "banana", "cherry", "orange"]

# Positive indexing
print(fruits[0])  # First element
print(fruits[2])  # Third element

# Negative indexing
print(fruits[-1])  # Last element
print(fruits[-2])  # Second-to-last element

# Lists are mutable, so an element can be replaced.
fruits[1] = "blueberry"
print(fruits)
````

### Illustrative Example

Suppose a list contains model training metrics. Positive and negative indexing can be useful when accessing the first or most recent value.

````python
loss_history = [0.91, 0.72, 0.55, 0.41, 0.33]

print(f"Initial loss: {loss_history[0]}")
print(f"Latest loss: {loss_history[-1]}")
print(f"Previous loss: {loss_history[-2]}")
````

### Edge Cases

Accessing an index outside the list raises `IndexError`.

```python
numbers = [10, 20, 30]

# numbers[3]   # IndexError
# numbers[-4]  # IndexError
```

The empty list has no valid indexes:

```python
empty = []

# empty[0]  # IndexError
```

## 2. Complete Slicing Syntax: `start`, `stop`, `step`

List slicing extracts a portion of a list.

The complete syntax is:

```python
list[start:stop:step]
```

Important:

- `start` is included.
- `stop` is excluded.
- `step` controls the distance between selected elements.
- Any of the three values can be omitted.

Examples:

```python
numbers[1:4]    # indexes 1, 2, 3
numbers[:3]     # beginning through index 2
numbers[2:]     # index 2 through the end
numbers[::2]    # every second element
numbers[::-1]   # reversed list
```

````python
numbers = [0, 1, 2, 3, 4, 5, 6, 7]

print(numbers[1:5])    # Start at 1, stop before 5
print(numbers[:4])     # First four elements
print(numbers[4:])     # From index 4 to the end
print(numbers[:])      # Entire list
print(numbers[::2])    # Every second element
print(numbers[1::2])   # Odd-indexed elements
print(numbers[::-1])   # Reverse the list
print(numbers[6:1:-1]) # Reverse from index 6 down to 2
````

### Illustrative Example

Slicing is useful when selecting batches or windows from sequential data.

````python
data = [10, 20, 30, 40, 50, 60, 70, 80]

# Select the first four values
first_batch = data[:4]

# Select every second value
sampled = data[::2]

# Reverse the sequence
reversed_data = data[::-1]

print("First batch:", first_batch)
print("Sampled:", sampled)
print("Reversed:", reversed_data)
````

### Edge Cases

A slice usually does not raise `IndexError` when the bounds extend beyond the list.

```python
numbers = [1, 2, 3]

print(numbers[0:100])  # [1, 2, 3]
```

A slice with a step of zero is invalid:

```python
# numbers[::0]  # ValueError: slice step cannot be zero
```

For a negative step, the start and stop positions normally need to describe a backward direction to produce elements.

## 3. List Comprehension

A list comprehension is a compact way to create a new list from an iterable.

Basic syntax:

```python
[expression for item in iterable]
```

A condition can be added:

```python
[expression for item in iterable if condition]
```

It is often equivalent to a `for` loop that builds a list, but it can be more concise.

````python
numbers = [1, 2, 3, 4, 5]

# Create a list containing squares
squares = [number ** 2 for number in numbers]

# Create a list containing only even numbers
even_numbers = [number for number in numbers if number % 2 == 0]

# Transform strings
names = ["alice", "bob", "charlie"]
uppercase_names = [name.upper() for name in names]

print(squares)
print(even_numbers)
print(uppercase_names)
````

### Illustrative Example

A common data-processing pattern is cleaning or transforming values before further processing.

````python
raw_scores = [72, 85, 91, 64, 88]

# Convert scores from percentages to values between 0 and 1
normalized_scores = [score / 100 for score in raw_scores]

print(normalized_scores)
````

### Edge Cases

A comprehension can contain nested loops, conditions, and conditional expressions, but excessive complexity can reduce readability.

```python
numbers = [1, 2, 3, 4]

result = ["even" if n % 2 == 0 else "odd" for n in numbers]
print(result)
```

For complicated multi-step logic, a normal `for` loop may be clearer.

## 4. Nested Lists and Accessing Nested Elements

A list can contain other lists. These are commonly called nested lists.

For example:

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
```

Accessing a nested element requires multiple indexes:

```python
matrix[row][column]
```

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matrix[0])       # First row
print(matrix[1][2])    # Second row, third element

# Modify a nested element
matrix[2][0] = 100

print(matrix)
````

### Illustrative Example

Nested lists can represent tabular data.

````python
students = [
    ["Alice", 85, 90],
    ["Bob", 78, 82],
    ["Charlie", 92, 95]
]

# Access Charlie's second score
print(students[2][2])

# Update Bob's first score
students[1][1] = 80

print(students)
````

### Edge Cases

Nested structures can have different lengths:

```python
data = [
    [1, 2],
    [3, 4, 5]
]
```

Code should not assume every inner list has the same length unless that is guaranteed.

Also, modifying a nested mutable object can affect other references to that object. This becomes especially important when copying nested lists.

## 5. Copying Lists: Assignment vs Shallow Copy

There is an important difference between assigning a list and copying it.

### Assignment

```python
a = [1, 2, 3]
b = a
```

`a` and `b` refer to the same list.

### Shallow Copy

A shallow copy creates a new outer list:

```python
b = a.copy()
```

or:

```python
b = a[:]
```

The outer list is independent, but nested mutable objects are still shared.

````python
# Assignment: both names refer to the same list
original = [1, 2, 3]
alias = original

alias.append(4)

print("original:", original)
print("alias:", alias)

# Shallow copy: creates a new outer list
original = [1, 2, 3]
copied = original.copy()

copied.append(4)

print("original:", original)
print("copied:", copied)
````

### Illustrative Example

A shallow copy works as expected for a flat list of immutable values.

````python
features = ["age", "income", "experience"]

# Create an independent outer list
selected_features = features.copy()

selected_features.append("education")

print("Original:", features)
print("Copy:", selected_features)
````

### Edge Cases

A shallow copy does not recursively copy nested mutable objects.

````python
original = [[1, 2], [3, 4]]
shallow_copy = original.copy()

# The inner list is shared
shallow_copy[0][0] = 99

print("Original:", original)
print("Shallow copy:", shallow_copy)
````

In this example, changing `shallow_copy[0][0]` also changes the corresponding inner list inside `original`.

## 6. Deep Copy

A deep copy recursively copies nested objects so that nested mutable structures are independent.

Python's `copy` module provides `deepcopy()`:

```python
from copy import deepcopy
```

Use deep copying when you need an independent copy of a nested mutable structure.

````python
from copy import deepcopy

original = [
    [1, 2],
    [3, 4]
]

deep_copy = deepcopy(original)

# Modify a nested object in the copy
deep_copy[0][0] = 99

print("Original:", original)
print("Deep copy:", deep_copy)
````

### Illustrative Example

Consider a nested configuration that needs to be modified independently.

````python
from copy import deepcopy

default_config = {
    "model": {
        "layers": [64, 32]
    }
}

# This example uses deepcopy because the structure is nested and mutable.
custom_config = deepcopy(default_config)

custom_config["model"]["layers"].append(16)

print("Default:", default_config)
print("Custom:", custom_config)
````

### Edge Cases

`deepcopy()` recursively copies objects, but not every object can or should be meaningfully deep-copied.

For ordinary nested lists, dictionaries, and other common mutable containers, it is generally appropriate.

Also note that deep copying can be more expensive than a shallow copy for large or deeply nested structures.

## 7. `index()`, `count()`, and `reverse()`

Lists provide several useful methods.

### `index()`

Returns the index of the first matching value.

```python
numbers.index(20)
```

### `count()`

Counts how many times a value occurs.

```python
numbers.count(20)
```

### `reverse()`

Reverses the list **in place** and returns `None`.

````python
numbers = [10, 20, 30, 20, 40, 20]

# Find the first occurrence
print(numbers.index(20))

# Count occurrences
print(numbers.count(20))

# Reverse the list in place
result = numbers.reverse()

print(numbers)
print(result)  # None
````

### Illustrative Example

These methods can be useful when analyzing a simple sequence of values.

````python
predictions = ["cat", "dog", "cat", "bird", "cat"]

cat_count = predictions.count("cat")
first_cat_index = predictions.index("cat")

print("Cat predictions:", cat_count)
print("First cat index:", first_cat_index)
````

### Edge Cases

`index()` raises `ValueError` when the value is not present.

```python
numbers = [1, 2, 3]

# numbers.index(10)  # ValueError
```

You can provide optional start and stop positions:

```python
numbers = [1, 2, 3, 2, 4]

print(numbers.index(2, 2))  # Search starting at index 2
```

`reverse()` modifies the existing list and returns `None`. Do not write:

```python
# reversed_list = numbers.reverse()
```

if you expect `reversed_list` to contain the reversed list.

## 8. `sort()` vs `sorted()`

Both `sort()` and `sorted()` can order values, but they behave differently.

### `list.sort()`

- Works on a list.
- Modifies the list in place.
- Returns `None`.

### `sorted()`

- Accepts any iterable.
- Creates and returns a new list.
- Leaves the original iterable unchanged.

Both support `key` and `reverse`.

````python
numbers = [5, 2, 8, 1, 3]

# sort() modifies the original list
numbers.sort()

print("After sort():", numbers)

numbers = [5, 2, 8, 1, 3]

# sorted() creates a new list
sorted_numbers = sorted(numbers)

print("Original:", numbers)
print("Sorted copy:", sorted_numbers)

# Descending order
numbers.sort(reverse=True)
print("Descending:", numbers)
````

### Illustrative Example

Sorting objects by a particular attribute or value is a common use case for `key`.

````python
students = [
    ("Alice", 88),
    ("Bob", 95),
    ("Charlie", 82)
]

# Sort by the score (second item in each tuple)
students_by_score = sorted(students, key=lambda student: student[1])

print(students_by_score)

# Sort from highest score to lowest
students_descending = sorted(
    students,
    key=lambda student: student[1],
    reverse=True
)

print(students_descending)
````

### Edge Cases

By default, sorting requires the values to be mutually comparable.

Mixing unrelated types can raise `TypeError`:

```python
# values = [1, "2", 3]
# sorted(values)  # TypeError
```

For strings, sorting is lexicographic and case-sensitive by default:

```python
words = ["banana", "Apple", "cherry"]
print(sorted(words))
```

A key function can provide custom ordering:

```python
print(sorted(words, key=str.lower))
```

## 9. `enumerate()`

`enumerate()` allows you to iterate over an iterable while keeping track of the index.

Instead of:

```python
for i in range(len(items)):
    print(i, items[i])
```

you can write:

```python
for i, item in enumerate(items):
    print(i, item)
```

The optional `start` argument controls the initial index.

````python
fruits = ["apple", "banana", "cherry"]

for index, fruit in enumerate(fruits):
    print(index, fruit)

# Start counting from 1 instead of 0
for number, fruit in enumerate(fruits, start=1):
    print(number, fruit)
````

### Illustrative Example

`enumerate()` is useful when displaying numbered results.

````python
models = ["Linear Regression", "Decision Tree", "Neural Network"]

for position, model in enumerate(models, start=1):
    print(f"{position}. {model}")
````

### Edge Cases

`enumerate()` returns an iterator, so it is lazy rather than immediately creating a list of pairs.

```python
items = ["a", "b", "c"]

enumerated = enumerate(items)

print(enumerated)
print(list(enumerated))
```

The iterator is consumed when iterated over.

## 10. `zip()`

`zip()` combines elements from multiple iterables position by position.

For example:

```python
names = ["Alice", "Bob", "Charlie"]
scores = [90, 85, 95]

zip(names, scores)
```

produces pairs:

```text
("Alice", 90)
("Bob", 85)
("Charlie", 95)
```

`zip()` is lazy and returns an iterator.

````python
names = ["Alice", "Bob", "Charlie"]
scores = [90, 85, 95]

for name, score in zip(names, scores):
    print(f"{name}: {score}")

# Convert the result into a list
pairs = list(zip(names, scores))
print(pairs)
````

### Illustrative Example

`zip()` is useful for combining related columns of data.

````python
features = ["age", "income", "experience"]
values = [25, 50000, 3]

record = dict(zip(features, values))

print(record)
````

### Edge Cases

By default, `zip()` stops when the shortest iterable is exhausted.

```python
a = [1, 2, 3]
b = ["x", "y"]

print(list(zip(a, b)))
# [(1, 'x'), (2, 'y')]
```

If you need to continue until the longest iterable is exhausted, `itertools.zip_longest()` can be used.

## 11. List Unpacking

List unpacking assigns elements from an iterable to multiple variables.

```python
numbers = [10, 20, 30]

a, b, c = numbers
```

The number of variables normally needs to match the number of elements.

Extended unpacking uses `*` to collect multiple values.

````python
numbers = [10, 20, 30]

# Basic unpacking
a, b, c = numbers

print(a)
print(b)
print(c)

# Extended unpacking
first, *middle, last = [1, 2, 3, 4, 5]

print("First:", first)
print("Middle:", middle)
print("Last:", last)

# The starred variable collects the remaining values into a list
head, *tail = [10, 20, 30, 40]

print("Head:", head)
print("Tail:", tail)
````

### Illustrative Example

Unpacking is useful when a function or operation returns several related values.

````python
data = ["Alice", 25, "Engineer"]

name, age, profession = data

print(f"Name: {name}")
print(f"Age: {age}")
print(f"Profession: {profession}")
````

### Edge Cases

Too few or too many values without extended unpacking causes `ValueError`.

```python
# a, b = [1, 2, 3]  # ValueError
# a, b, c = [1, 2]   # ValueError
```

With `*`, the collected portion can contain zero or more values:

```python
first, *rest = [10]
print(first)  # 10
print(rest)   # []
```

Only one starred target can be used in a single unpacking assignment.

## 12. Aliasing and Mutable References

Python variables are names bound to objects.

When two variables refer to the same mutable list, they are aliases:

```python
a = [1, 2, 3]
b = a
```

There is still only one list object.

Changing the list through either name changes the same object.

This is different from making a copy.

````python
a = [1, 2, 3]
b = a

print(a is b)  # True

# Modify the list through b
b.append(4)

# Both names see the same change
print("a:", a)
print("b:", b)
````

### Illustrative Example

Aliasing can occur when a list is passed to a function. The function receives a reference to the same list object, so mutating the list affects the caller's list.

````python
def add_item(items, item):
    # This mutates the existing list object.
    items.append(item)

numbers = [1, 2, 3]

add_item(numbers, 4)

print(numbers)
````

### Edge Cases

Reassignment and mutation are different operations.

Reassignment changes what a variable refers to:

```python
a = [1, 2, 3]
b = a

b = [4, 5, 6]

print(a)  # [1, 2, 3]
print(b)  # [4, 5, 6]
```

Mutation changes the existing object:

```python
a = [1, 2, 3]
b = a

b.append(4)

print(a)  # [1, 2, 3, 4]
print(b)  # [1, 2, 3, 4]
```

This distinction becomes especially important with nested lists, function arguments, and shallow/deep copying.

# Combined Example

The following example combines indexing, slicing, comprehensions, `enumerate()`, `zip()`, sorting, and copying in one small workflow.

````python
# Original data
names = ["Alice", "Bob", "Charlie", "Diana"]
scores = [88, 72, 95, 81]

# Combine related data
records = list(zip(names, scores))

# Sort records by score from highest to lowest
records_sorted = sorted(records, key=lambda record: record[1], reverse=True)

# Create a list of students who scored at least 80
high_scorers = [
    name
    for name, score in records_sorted
    if score >= 80
]

# Display the ranked records
for rank, (name, score) in enumerate(records_sorted, start=1):
    print(f"{rank}. {name}: {score}")

print("High scorers:", high_scorers)

# Copy the list of records
records_copy = records_sorted.copy()

# Modify the outer list without changing the original ordering list
records_copy.append(("Eve", 90))

print("Original records:", records_sorted)
print("Copied records:", records_copy)
````

# Common List Pitfalls

### 1. Confusing assignment with copying

```python
a = [1, 2, 3]
b = a
```

This creates an alias, not a copy.

### 2. Forgetting that `sort()` and `reverse()` mutate the list

```python
items.sort()
items.reverse()
```

Both return `None`.

### 3. Forgetting that `sorted()` creates a new list

```python
new_items = sorted(items)
```

The original list remains unchanged.

### 4. Assuming `.copy()` is always enough

For nested mutable structures, `.copy()` is shallow.

### 5. Assuming `zip()` pads shorter iterables

Normal `zip()` stops at the shortest iterable.

### 6. Modifying a list while iterating over it

This can produce skipped elements or unexpected behavior. If elements need to be removed conditionally, it is often safer to build a new list or iterate over a copy.

Example:

```python
numbers = [1, 2, 3, 4, 5]

# Safer approach: create a new list
numbers = [n for n in numbers if n % 2 != 0]

print(numbers)
```
