# Lambda Functions

A reference notebook covering Python lambda expressions, parameters and arguments, default arguments, conditional expressions, sorting, `map()`, `filter()`, higher-order functions, and functions as objects.

A lambda is an anonymous function expression that contains a single expression and returns its result.

## 1. Normal Functions vs Lambda Functions

A normal function is defined with `def` and can contain multiple statements.

A lambda is an expression that creates a function object inline.

### Normal function

```python
def square(x):
    return x ** 2
```

### Equivalent lambda

```python
square = lambda x: x ** 2
```

Lambdas are especially useful for short operations passed directly to another function.

````python
def square_function(x):
    return x ** 2

square_lambda = lambda x: x ** 2

print(square_function(5))
print(square_lambda(5))

print(type(square_lambda))
````

### Edge cases

A lambda cannot contain arbitrary statements such as assignments, `for` statements, or `try` blocks. It contains a single expression.

The expression can still contain function calls, conditional expressions, comprehensions, and other expression-level constructs.

````python
# Valid: function call inside a lambda
get_length = lambda text: len(text)

print(get_length("Python"))

# Invalid syntax:
# bad = lambda x:
#     y = x + 1
#     return y
````

## 2. Lambda Syntax

The general syntax is:

```python
lambda parameters: expression
```

The value of the expression becomes the function's return value.

````python
add_five = lambda x: x + 5
multiply = lambda x, y: x * y

print(add_five(10))
print(multiply(4, 6))
````

### Practical example

````python
temperatures_c = [0, 10, 20, 30]

to_fahrenheit = lambda c: (c * 9 / 5) + 32

temperatures_f = [to_fahrenheit(c) for c in temperatures_c]

print(temperatures_f)
````

### Important point

Lambdas do not have a special runtime type. A lambda creates an ordinary function object.

````python
operation = lambda x: x * 2

print(type(operation))
print(operation.__name__)  # Usually '<lambda>'
````

## 3. Parameters and Arguments

Lambda functions support positional parameters, keyword arguments, default values, `*args`, and `**kwargs` just like normal functions.

````python
# One parameter
double = lambda x: x * 2

# Multiple parameters
add = lambda x, y: x + y

# Calling with positional arguments
print(add(10, 20))

# Calling with keyword arguments
print(add(x=10, y=20))
````

### Parameter vs argument

- A **parameter** is the name defined in the function.
- An **argument** is the value supplied when calling the function.

````python
combine = lambda first, second: f"{first}-{second}"

# first and second are parameters
# "A" and "B" are arguments
print(combine("A", "B"))
````

## 4. Default Arguments

Lambdas can define default parameter values just like normal functions.

````python
power = lambda x, exponent=2: x ** exponent

print(power(5))
print(power(5, 3))

print(power(x=2, exponent=4))
````

### Practical example

````python
format_score = lambda score, precision=2: f"{score:.{precision}f}"

print(format_score(0.93456))
print(format_score(0.93456, 4))
````

### Edge cases

Default argument values are evaluated when the function is created, following the same default-argument behavior as normal functions.

````python
default_value = 10

get_value = lambda x=default_value: x

default_value = 20

print(get_value())  # 10
````

## 5. Multiple Arguments

A lambda can accept multiple parameters separated by commas.

````python
average = lambda a, b, c: (a + b + c) / 3

print(average(10, 20, 30))

create_record = lambda name, age: {
    "name": name,
    "age": age
}

print(create_record("Alice", 21))
````

### Edge cases

The number of supplied positional arguments must match the number of required parameters, unless defaults or variadic parameters are used.

````python
# add = lambda x, y: x + y
# add(10)  # TypeError
````

## 6. Conditional Expressions in Lambdas

Although lambdas cannot contain an `if` statement, they can use Python's conditional expression:

```python
value_if_true if condition else value_if_false
```

````python
absolute_value = lambda x: x if x >= 0 else -x

print(absolute_value(10))
print(absolute_value(-10))
````

### Practical example

````python
classify_score = lambda score: (
    "excellent" if score >= 90
    else "good" if score >= 70
    else "needs improvement"
)

print(classify_score(95))
print(classify_score(75))
print(classify_score(50))
````

### Edge cases

Nested conditional expressions are valid but can become difficult to read. For complex decision logic, a normal `def` function is usually clearer.

## 7. Lambda with `sorted()` and `key=`

The `key` argument of `sorted()` accepts a callable that determines the comparison value for each element.

Lambdas are commonly used when the sorting rule is needed only once.

````python
students = [
    {"name": "Alice", "score": 85},
    {"name": "Bob", "score": 92},
    {"name": "Charlie", "score": 78}
]

# Sort by score
by_score = sorted(students, key=lambda student: student["score"])

print(by_score)

# Sort by score in descending order
by_score_desc = sorted(
    students,
    key=lambda student: student["score"],
    reverse=True
)

print(by_score_desc)
````

### Practical example: sorting strings

````python
words = ["Python", "AI", "Machine Learning", "ML"]

# Sort by length
sorted_words = sorted(words, key=lambda word: len(word))

print(sorted_words)
````

### Sorting by multiple criteria

````python
records = [
    ("Alice", 90),
    ("Bob", 80),
    ("Charlie", 90),
    ("David", 80)
]

# First sort by score, then by name
result = sorted(records, key=lambda record: (record[1], record[0]))

print(result)
````

### Edge cases

`key=` expects a callable. The key function is applied to each item and the resulting values are used for sorting.

## 8. Lambda with `map()`

`map(function, iterable)` applies a callable to every item and returns a lazy iterator.

A lambda is often used when the transformation is short and used only once.

````python
numbers = [1, 2, 3, 4, 5]

squares = map(lambda x: x ** 2, numbers)

print(squares)
print(list(squares))
````

### Practical example

````python
prices = [100, 250, 500]

discounted = map(lambda price: price * 0.9, prices)

print(list(discounted))
````

### Important point

`map()` returns an iterator in Python 3. It is consumed as it is iterated.

````python
numbers = [1, 2, 3]

result = map(lambda x: x * 2, numbers)

print(list(result))
print(list(result))  # Empty: iterator was already consumed
````

## 9. Lambda with `filter()`

`filter(function, iterable)` keeps elements for which the callable returns a truthy value.

A lambda is useful for short filtering conditions.

````python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = filter(lambda x: x % 2 == 0, numbers)

print(list(even_numbers))
````

### Practical example

````python
users = [
    {"name": "Alice", "active": True},
    {"name": "Bob", "active": False},
    {"name": "Charlie", "active": True}
]

active_users = filter(lambda user: user["active"], users)

print(list(active_users))
````

### Edge cases

If the filtering function is `None`, `filter()` keeps truthy elements directly.

````python
values = [0, 1, "", "Python", None, [], [1]]

print(list(filter(None, values)))
````

## 10. Higher-Order Functions

A higher-order function is a function that:
- accepts another function as an argument, and/or
- returns a function

Because functions are objects in Python, they can be stored in variables, passed around, and returned.

````python
def apply_operation(value, operation):
    return operation(value)

double = lambda x: x * 2
square = lambda x: x ** 2

print(apply_operation(5, double))
print(apply_operation(5, square))
````

### Practical example

````python
def transform_values(values, operation):
    return [operation(value) for value in values]

numbers = [1, 2, 3, 4]

print(transform_values(numbers, lambda x: x * 10))
print(transform_values(numbers, lambda x: x ** 2))
````

### Returning a function

````python
def make_multiplier(factor):
    return lambda value: value * factor

double = make_multiplier(2)
triple = make_multiplier(3)

print(double(10))
print(triple(10))
````

## 11. Functions as Objects

Functions are first-class objects in Python.

They can be:
- assigned to variables
- stored in lists or dictionaries
- passed as arguments
- returned from functions
- attached to other data structures

````python
def greet(name):
    return f"Hello, {name}!"

# Assign function object to another name
say_hello = greet

print(say_hello("Alice"))

# Store functions in a list
operations = [
    lambda x: x + 1,
    lambda x: x * 2,
    lambda x: x ** 2
]

for operation in operations:
    print(operation(5))
````

### Functions in dictionaries

A dictionary can map names to callable operations.

````python
operations = {
    "double": lambda x: x * 2,
    "square": lambda x: x ** 2,
    "negative": lambda x: -x
}

value = 5

for name, operation in operations.items():
    print(name, "->", operation(value))
````

## 12. Lambda with Multiple Iterables

`map()` can pass one element from each iterable to the callable.

The lambda therefore needs a parameter for each iterable.

````python
numbers = [1, 2, 3]
offsets = [10, 20, 30]

result = map(lambda number, offset: number + offset, numbers, offsets)

print(list(result))
````

### Edge cases

When multiple iterables have different lengths, `map()` stops when the shortest iterable is exhausted.

````python
a = [1, 2, 3]
b = [10, 20]

result = map(lambda x, y: x + y, a, b)

print(list(result))
````

## 13. Lambda with `reduce()`

`reduce()` is provided by `functools` and repeatedly combines values using a callable.

Syntax:

```python
reduce(function, iterable[, initializer])
```

Unlike `map()` and `filter()`, `reduce()` produces a single accumulated result.

````python
from functools import reduce

numbers = [1, 2, 3, 4]

total = reduce(lambda accumulator, value: accumulator + value, numbers)

print(total)
````

### With an initializer

````python
from functools import reduce

numbers = [1, 2, 3]

total = reduce(
    lambda accumulator, value: accumulator + value,
    numbers,
    100
)

print(total)
````

### Practical example

````python
from functools import reduce

numbers = [2, 3, 4]

product = reduce(lambda a, b: a * b, numbers)

print(product)
````

### Edge cases

Reducing an empty iterable without an initializer raises `TypeError`. Supplying an initializer provides the starting accumulator value.

````python
from functools import reduce

# reduce(lambda a, b: a + b, [])  # TypeError

print(reduce(lambda a, b: a + b, [], 0))
````

## 14. Lambda vs List Comprehension

For transformations, list comprehensions are often clearer than `map()` with a lambda.

These two examples produce equivalent lists.

````python
numbers = [1, 2, 3, 4]

result_map = list(map(lambda x: x ** 2, numbers))
result_comprehension = [x ** 2 for x in numbers]

print(result_map)
print(result_comprehension)
````

### Filtering comparison

````python
numbers = [1, 2, 3, 4, 5, 6]

result_filter = list(filter(lambda x: x % 2 == 0, numbers))
result_comprehension = [x for x in numbers if x % 2 == 0]

print(result_filter)
print(result_comprehension)
````

### General guideline

Use a lambda when it makes a short callback clearer. Use `def` when the operation needs multiple statements, documentation, debugging, reuse, or substantial logic.

## 15. Lambda Scope and Closures

A lambda can capture variables from its surrounding scope, just like a normal nested function.

The captured variable behavior follows Python's closure rules.

````python
def make_adder(amount):
    return lambda value: value + amount

add_10 = make_adder(10)
add_100 = make_adder(100)

print(add_10(5))
print(add_100(5))
````

### Late binding example

When lambdas are created inside a loop, closures capture the variable rather than taking a snapshot of its value at each iteration.

````python
functions = []

for i in range(3):
    functions.append(lambda: i)

print([function() for function in functions])  # [2, 2, 2]
````

### Capturing the current value with a default argument

````python
functions = []

for i in range(3):
    functions.append(lambda i=i: i)

print([function() for function in functions])  # [0, 1, 2]
````

## 16. Lambda and `sorted()` with Dictionaries

A common pattern is sorting dictionary data by a particular field.

````python
models = [
    {"name": "Model A", "accuracy": 0.91},
    {"name": "Model B", "accuracy": 0.95},
    {"name": "Model C", "accuracy": 0.89}
]

best_first = sorted(
    models,
    key=lambda model: model["accuracy"],
    reverse=True
)

for model in best_first:
    print(model)
````

## 17. Lambda as a Small Callback

Lambdas are useful as short callbacks where defining a named function would add unnecessary ceremony.

````python
numbers = [5, 1, 8, 2, 7]

# The lambda is used only as the sorting rule.
numbers.sort(key=lambda x: -x)

print(numbers)
````

### Edge cases

Do not force everything into a lambda. If a callback has a meaningful name or complicated logic, a named function usually improves readability and maintainability.

# Combined Example

This example combines lambdas, sorting, filtering, mapping, and a higher-order function.

````python
students = [
    {"name": "Alice", "score": 91},
    {"name": "Bob", "score": 67},
    {"name": "Charlie", "score": 84},
    {"name": "David", "score": 95}
]

# Keep students who passed
passed = filter(lambda student: student["score"] >= 70, students)

# Extract names
names = map(lambda student: student["name"], passed)

# Sort the names alphabetically
result = sorted(names)

print(result)

# Sort original students by score
ranked = sorted(
    students,
    key=lambda student: student["score"],
    reverse=True
)

print(ranked)
````

# Common Pitfalls

- A lambda contains one expression, not an arbitrary block of statements.
- `lambda` does not create a special type; it creates a normal function object.
- `map()` and `filter()` return lazy iterators in Python 3.
- Iterators are consumed after iteration.
- `reduce()` on an empty iterable requires an initializer.
- `sorted(..., key=...)` expects a callable.
- Complex lambda expressions can hurt readability.
- Use a normal `def` function when logic is substantial or deserves a meaningful name.
- Lambdas created in loops can exhibit late-binding behavior.
- Default lambda arguments are evaluated when the lambda is created.
