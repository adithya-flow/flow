# Collections

A reference notebook for Python's `collections` module, covering `Counter`, `defaultdict`, `deque`, `namedtuple`, `OrderedDict`, and `ChainMap`.

## 1. Importing from `collections`

These specialized containers complement Python's built-in collection types.

````python
from collections import Counter, defaultdict, deque, namedtuple, OrderedDict, ChainMap

print(Counter)
print(defaultdict)
print(deque)
print(namedtuple)
print(OrderedDict)
print(ChainMap)
````

# Counter

## 2. Counting items

`Counter` is a dictionary subclass designed for counting hashable objects.

````python
from collections import Counter

numbers = [1, 2, 2, 3, 3, 3, 4]
counts = Counter(numbers)

print(counts)
print(counts[2])
print(counts["missing"])  # Missing Counter keys return 0
````

## 3. Counting characters

````python
from collections import Counter

text = "banana"
counts = Counter(text)

print(counts)
print(counts["a"])
````

## 4. Counting words

````python
from collections import Counter

text = "python is useful and python is readable"
word_counts = Counter(text.split())

print(word_counts)
print(dict(word_counts))
````

## 5. Creating a Counter from a mapping

````python
from collections import Counter

counts = Counter({"apple": 3, "banana": 2, "orange": 1})
print(counts)
````

## 6. `most_common()`

Returns elements ordered by frequency. Pass `n` for the top `n`.

````python
from collections import Counter

counts = Counter("mississippi")

print(counts.most_common())
print(counts.most_common(2))
````

## 7. `update()`

Adds counts; it does not replace existing counts.

````python
from collections import Counter

counts = Counter(["a", "a", "b"])
counts.update(["a", "c"])
counts.update({"b": 2, "d": 4})

print(counts)
````

## 8. `subtract()`

Subtracts counts and may produce zero or negative values.

````python
from collections import Counter

counts = Counter({"a": 5, "b": 3})
counts.subtract({"a": 2, "b": 5, "c": 1})

print(counts)
````

## 9. `elements()` and `total()`

`elements()` repeats items according to positive counts. `total()` sums counts.

````python
from collections import Counter

counts = Counter({"a": 3, "b": 2, "c": 0, "d": -1})

print(list(counts.elements()))
print(counts.total())
````

## 10. Counter arithmetic

`+`, `-`, `&`, and `|` implement multiset-style operations.

````python
from collections import Counter

a = Counter({"apple": 3, "banana": 2})
b = Counter({"apple": 1, "banana": 4, "orange": 2})

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Intersection:", a & b)
print("Union:", a | b)
````

## 11. Removing non-positive counts

Unary `+` keeps only positive counts.

````python
from collections import Counter

counts = Counter({"a": 3, "b": 0, "c": -2})
print(counts)
print(+counts)
````

## 12. Counter as a frequency dictionary

A common use is converting raw categorical data into frequencies.

````python
from collections import Counter

labels = ["cat", "dog", "cat", "bird", "dog", "cat"]
frequency = Counter(labels)

print(frequency)
print(dict(frequency))
print(frequency.most_common())
````

## 13. Practical example: top-K categories

````python
from collections import Counter

labels = ["cat", "dog", "cat", "bird", "dog", "cat", "dog", "horse"]
top_2 = Counter(labels).most_common(2)

print(top_2)
````

## 14. Practical example: token frequencies

````python
from collections import Counter

sentence = "Python is simple. Python is powerful. Python is useful."
tokens = sentence.lower().replace(".", "").split()

frequencies = Counter(tokens)
print(frequencies)
print(frequencies.most_common(3))
````

## 15. Practical example: inventory counting

````python
from collections import Counter

inventory = Counter({"apple": 10, "banana": 6, "orange": 4})
purchase = Counter({"apple": 2, "banana": 3})

print("Remaining:", inventory - purchase)
````

## 16. Counter edge cases

Counter keys must be hashable. Missing keys return `0`, but a missing key is not necessarily stored. Counts can also be zero or negative.

````python
from collections import Counter

counts = Counter({"a": 0})

print(counts["a"])
print(counts["missing"])
print("a" in counts)
print("missing" in counts)

# Counter([[1, 2]])  # TypeError: lists are unhashable
````

# defaultdict

## 17. Automatic defaults

`defaultdict` calls its `default_factory` when a missing key is accessed.

````python
from collections import defaultdict

counts = defaultdict(int)

counts["apple"] += 1
counts["apple"] += 1
counts["banana"] += 1

print(dict(counts))
````

## 18. `defaultdict(list)` for grouping

````python
from collections import defaultdict

groups = defaultdict(list)

groups["fruit"].append("apple")
groups["fruit"].append("banana")
groups["vegetable"].append("carrot")

print(dict(groups))
````

## 19. `defaultdict(set)` for unique grouping

````python
from collections import defaultdict

groups = defaultdict(set)

groups["python"].add("beginner")
groups["python"].add("beginner")
groups["python"].add("data")

print(dict(groups))
````

## 20. `defaultdict` edge case

Accessing a missing key creates it. Use `get()` or membership testing when you do not want that side effect.

````python
from collections import defaultdict

data = defaultdict(list)

print(dict(data))
print(data["missing"])
print(dict(data))

other = defaultdict(list)
print(other.get("missing"))
print(dict(other))
````

# deque

## 21. Double-ended queues

`deque` supports efficient insertion and removal from both ends.

````python
from collections import deque

items = deque(["b", "c"])

items.append("d")
items.appendleft("a")

print(items)
print(items.pop())
print(items.popleft())
print(items)
````

## 22. `deque` as a queue

````python
from collections import deque

queue = deque(["task 1", "task 2", "task 3"])

while queue:
    task = queue.popleft()
    print("Processing:", task)
````

## 23. `deque` as a stack

````python
from collections import deque

stack = deque()

for value in ["A", "B", "C"]:
    stack.append(value)

while stack:
    print("Popped:", stack.pop())
````

## 24. `rotate()`

````python
from collections import deque

items = deque([1, 2, 3, 4])

items.rotate(1)
print(items)

items.rotate(-2)
print(items)
````

## 25. Bounded `deque`

`maxlen` keeps only the most recent values.

````python
from collections import deque

recent = deque(maxlen=3)

for value in [1, 2, 3, 4, 5]:
    recent.append(value)
    print(recent)
````

# namedtuple

## 26. Named immutable records

`namedtuple()` creates a tuple subclass whose fields can be accessed by name or position.

````python
from collections import namedtuple

Point = namedtuple("Point", ["x", "y"])
point = Point(10, 20)

print(point)
print(point.x)
print(point.y)
print(point[0])
````

## 27. `namedtuple` metadata and conversion

````python
from collections import namedtuple

User = namedtuple("User", ["name", "age"])
user = User("Jayaadithya", 21)

print(User._fields)
print(user._asdict())
````

## 28. Practical `namedtuple` example

````python
from collections import namedtuple

ModelResult = namedtuple("ModelResult", ["accuracy", "loss"])
result = ModelResult(0.94, 0.21)

print(result.accuracy)
print(result.loss)
````

# OrderedDict

## 29. What is `OrderedDict`?

Modern Python dictionaries preserve insertion order, so `OrderedDict` is no longer required merely for that purpose. It remains useful for explicit order manipulation.

````python
from collections import OrderedDict

data = OrderedDict([
    ("a", 1),
    ("b", 2),
    ("c", 3),
])

print(data)
````

## 30. `move_to_end()`

````python
from collections import OrderedDict

data = OrderedDict([("a", 1), ("b", 2), ("c", 3)])

data.move_to_end("a")
print(data)

data.move_to_end("c", last=False)
print(data)
````

## 31. `popitem()` with `OrderedDict`

````python
from collections import OrderedDict

data = OrderedDict([("a", 1), ("b", 2), ("c", 3)])

print(data.popitem())
print(data.popitem(last=False))
print(data)
````

# ChainMap

## 32. Layered mappings

`ChainMap` provides a single lookup view over multiple mappings. Lookup starts with the first mapping.

````python
from collections import ChainMap

defaults = {"color": "blue", "size": 10}
user = {"color": "red"}

config = ChainMap(user, defaults)

print(config["color"])
print(config["size"])
````

## 33. `ChainMap` lookup precedence

````python
from collections import ChainMap

first = {"x": 1, "shared": "first"}
second = {"y": 2, "shared": "second"}

combined = ChainMap(first, second)

print(combined["shared"])

combined["x"] = 100
combined["z"] = 300

print(first)
print(second)
````

## 34. `maps`, `new_child()`, and `parents`

````python
from collections import ChainMap

base = ChainMap({"mode": "production", "debug": False})

temporary = base.new_child({"debug": True})

print(temporary["debug"])
print(temporary.maps)
print(temporary.parents)
````

## 35. Choosing the right collection

| Type | Main purpose |
|---|---|
| `Counter` | Frequency counting |
| `defaultdict` | Automatic defaults |
| `deque` | Efficient double-ended queue |
| `namedtuple` | Named immutable records |
| `OrderedDict` | Explicit order operations |
| `ChainMap` | Layered mapping lookup |

````python
from collections import Counter, defaultdict, deque, namedtuple, OrderedDict, ChainMap

print("Counter     -> counting")
print("defaultdict -> automatic defaults")
print("deque       -> both-end queue operations")
print("namedtuple  -> named immutable records")
print("OrderedDict -> order manipulation")
print("ChainMap    -> layered mappings")
````

## 36. Common pitfalls

1. `Counter` keys must be hashable.
2. Missing Counter keys return `0`.
3. Accessing a missing `defaultdict` key creates it.
4. Empty deque removal raises `IndexError`.
5. A bounded deque discards items from the opposite end when full.
6. `namedtuple` instances are immutable.
7. `OrderedDict` is usually unnecessary just to preserve insertion order in modern Python.
8. `ChainMap` layers mappings; it does not create a merged dictionary.

````python
from collections import Counter, defaultdict, deque

counts = Counter(["a", "b", "a"])
print(counts)

groups = defaultdict(list)
groups["new"].append(1)
print(groups)

items = deque()
try:
    items.popleft()
except IndexError:
    print("Cannot remove from an empty deque.")
````
