# Map, Filter, Reduce

A reference notebook covering `map()`, `filter()`, `reduce()`, iterables, iterators, `iter()`, `next()`, lazy evaluation, generator expressions, and comparisons with list comprehensions.

These tools are especially useful for expressing transformations, filtering data, and combining sequences into accumulated results.

## 1. `map()` with Normal Functions

`map(function, iterable)` applies a callable to every item in an iterable.

In Python 3, `map()` returns a lazy iterator rather than a list.

````python
def square(x):
    return x ** 2

numbers = [1, 2, 3, 4, 5]

result = map(square, numbers)

print(result)
print(list(result))
````

### Practical example

````python
def normalize_name(name):
    return name.strip().title()

names = [" alice ", "BOB", "charlie "]

normalized = map(normalize_name, names)

print(list(normalized))
````

### Edge cases

The function supplied to `map()` must accept the number of arguments provided by the iterables. The result is lazy and is consumed during iteration.

````python
def double(x):
    return x * 2

numbers = [1, 2, 3]

result = map(double, numbers)

print(list(result))
print(list(result))  # Empty: iterator already consumed
````

## 2. `map()` with Multiple Iterables

`map()` can accept multiple iterables.

The function receives one item from each iterable on every iteration.

````python
def add(a, b):
    return a + b

first = [1, 2, 3]
second = [10, 20, 30]

result = map(add, first, second)

print(list(result))
````

### Practical example

````python
prices = [100, 200, 300]
quantities = [2, 3, 4]

totals = map(
    lambda price, quantity: price * quantity,
    prices,
    quantities
)

print(list(totals))
````

### Edge case: different lengths

When multiple iterables have different lengths, `map()` stops when the shortest iterable is exhausted.

````python
a = [1, 2, 3, 4]
b = [10, 20]

result = map(lambda x, y: x + y, a, b)

print(list(result))
````

## 3. `map()` with Lambda

Lambdas are commonly used with `map()` for short transformations.

````python
numbers = [1, 2, 3, 4, 5]

squares = map(lambda x: x ** 2, numbers)

print(list(squares))
````

### Practical example

````python
temperatures_c = [0, 10, 20, 30]

temperatures_f = map(
    lambda c: (c * 9 / 5) + 32,
    temperatures_c
)

print(list(temperatures_f))
````

### Equivalent list comprehension

````python
numbers = [1, 2, 3, 4]

mapped = list(map(lambda x: x * 10, numbers))
comprehension = [x * 10 for x in numbers]

print(mapped)
print(comprehension)
````

## 4. `filter()` with Normal Functions

`filter(function, iterable)` keeps elements for which the function returns a truthy value.

Like `map()`, `filter()` returns a lazy iterator in Python 3.

````python
def is_even(number):
    return number % 2 == 0

numbers = [1, 2, 3, 4, 5, 6]

result = filter(is_even, numbers)

print(result)
print(list(result))
````

### Practical example

````python
def is_valid_score(score):
    return 0 <= score <= 100

scores = [95, -5, 72, 110, 88]

valid_scores = filter(is_valid_score, scores)

print(list(valid_scores))
````

### Filtering objects

````python
def is_active(user):
    return user["active"]

users = [
    {"name": "Alice", "active": True},
    {"name": "Bob", "active": False},
    {"name": "Charlie", "active": True}
]

active_users = filter(is_active, users)

print(list(active_users))
````

## 5. `filter()` with Lambda

For short predicates, a lambda is often convenient.

````python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = filter(lambda x: x % 2 == 0, numbers)

print(list(even_numbers))
````

### Practical example

````python
words = ["AI", "Python", "ML", "Data Science", "API"]

long_words = filter(lambda word: len(word) > 3, words)

print(list(long_words))
````

### Filtering truthy values with `None`

If the function argument is `None`, `filter()` keeps truthy elements.

````python
values = [0, 1, "", "Python", None, [], [1], False, True]

print(list(filter(None, values)))
````

## 6. `reduce()` from `functools`

`reduce()` repeatedly applies a function to an iterable and reduces it to a single accumulated result.

It is imported from `functools`.

```python
from functools import reduce
```

Conceptually:

```text
[1, 2, 3, 4]
  ↓
((1 + 2) + 3) + 4
  ↓
10
```

````python
from functools import reduce

numbers = [1, 2, 3, 4]

total = reduce(lambda accumulator, value: accumulator + value, numbers)

print(total)
````

### Practical example: product

````python
from functools import reduce

numbers = [2, 3, 4, 5]

product = reduce(lambda a, b: a * b, numbers)

print(product)
````

### Practical example: maximum

````python
from functools import reduce

numbers = [12, 7, 25, 4, 18]

maximum = reduce(lambda a, b: a if a > b else b, numbers)

print(maximum)
````

## 7. `reduce()` with an Initializer

The optional third argument provides the initial accumulator value.

This is useful for empty iterables and when the calculation should begin from a specific value.

````python
from functools import reduce

numbers = [1, 2, 3]

total = reduce(
    lambda accumulator, value: accumulator + value,
    numbers,
    100
)

print(total)  # 106
````

### Empty iterable

Without an initializer, reducing an empty iterable raises `TypeError`. With an initializer, the initializer becomes the result when there are no input values.

````python
from functools import reduce

# reduce(lambda a, b: a + b, [])  # TypeError

print(reduce(lambda a, b: a + b, [], 0))
print(reduce(lambda a, b: a + b, [], 100))
````

## 8. `map()` vs `filter()` vs `reduce()`

| Function | Main purpose | Output |
|---|---|---|
| `map()` | Transform every element | Iterator |
| `filter()` | Keep selected elements | Iterator |
| `reduce()` | Accumulate elements | Single value |

Examples:

```python
map(f, data)
filter(predicate, data)
reduce(combine, data)
```

````python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

mapped = map(lambda x: x * 2, numbers)
filtered = filter(lambda x: x % 2 == 0, numbers)
reduced = reduce(lambda a, b: a + b, numbers)

print("map:", list(mapped))
print("filter:", list(filtered))
print("reduce:", reduced)
````

## 9. Combining `map()`, `filter()`, and `reduce()`

These functions can be chained to build a processing pipeline.

````python
from functools import reduce

numbers = [1, 2, 3, 4, 5, 6]

# 1. Keep even numbers
even_numbers = filter(lambda x: x % 2 == 0, numbers)

# 2. Square them
squared = map(lambda x: x ** 2, even_numbers)

# 3. Add the results
total = reduce(lambda a, b: a + b, squared)

print(total)
````

### Equivalent step-by-step version

Keeping intermediate iterators explicit can make debugging easier.

````python
from functools import reduce

numbers = [1, 2, 3, 4, 5, 6]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
squared = list(map(lambda x: x ** 2, even_numbers))
total = reduce(lambda a, b: a + b, squared)

print("Even:", even_numbers)
print("Squared:", squared)
print("Total:", total)
````

## 10. Iterable vs Iterator

An **iterable** is an object that can return an iterator.

Examples:
- lists
- tuples
- strings
- dictionaries
- sets
- ranges

An **iterator** is an object that produces values one at a time and keeps track of its current position.

Typical iterator operations:
- `iter()`
- `next()`

````python
numbers = [10, 20, 30]

print(hasattr(numbers, "__iter__"))
print(hasattr(numbers, "__next__"))

iterator = iter(numbers)

print(hasattr(iterator, "__iter__"))
print(hasattr(iterator, "__next__"))
````

### Important distinction

A list is iterable but is not itself an iterator.

An iterator implements `__next__()` and therefore can produce its next value directly.

## 11. `iter()`

`iter(iterable)` obtains an iterator from an iterable.

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
````

### Practical example

````python
text = "Python"

characters = iter(text)

print(next(characters))
print(next(characters))
print(next(characters))
````

### Edge cases

Calling `iter()` on an object that is not iterable raises `TypeError`.

````python
# iter(42)  # TypeError: int is not iterable
````

## 12. `next()`

`next(iterator)` retrieves the next item from an iterator.

When no items remain, it raises `StopIteration`.

````python
numbers = [10, 20]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))

# next(iterator)  # StopIteration
````

### Providing a default value

`next(iterator, default)` returns the default instead of raising `StopIteration` when the iterator is exhausted.

````python
numbers = [10]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator, "No more values"))
print(next(iterator, "No more values"))
````

## 13. Iterating with `for`

A `for` loop uses the iterator protocol internally.

Conceptually, Python obtains an iterator and repeatedly calls `next()` until `StopIteration` occurs.

````python
numbers = [10, 20, 30]

for number in numbers:
    print(number)
````

### Rough conceptual equivalent

The following illustrates the basic iterator mechanism behind a `for` loop.

````python
numbers = [10, 20, 30]
iterator = iter(numbers)

while True:
    try:
        value = next(iterator)
        print(value)
    except StopIteration:
        break
````

## 14. Lazy Evaluation

Lazy evaluation means values are produced only when they are requested rather than all being computed immediately.

`map()`, `filter()`, `range()`, and generators are common examples of lazy or iterator-based behavior.

````python
numbers = [1, 2, 3, 4, 5]

mapped = map(lambda x: x * 10, numbers)

print(mapped)

# Values are produced when consumed
for value in mapped:
    print(value)
````

### Why laziness matters

Lazy processing can:
- avoid creating unnecessary intermediate collections
- reduce memory usage
- allow processing of large or even unbounded streams
- delay computation until results are needed

````python
# range does not create a list of one billion integers
large_range = range(1_000_000_000)

print(large_range)
print(next(iter(large_range)))
````

### Edge cases

Lazy iterators are single-pass in the normal iterator model. If you need to reuse the results, materialize them into a collection such as `list()` when appropriate.

## 15. Generator Expressions

A generator expression is similar to a list comprehension but uses parentheses instead of brackets.

```python
(expression for item in iterable)
```

It produces a generator object and evaluates values lazily.

````python
numbers = [1, 2, 3, 4, 5]

squares_generator = (x ** 2 for x in numbers)

print(squares_generator)
print(next(squares_generator))
print(next(squares_generator))

print(list(squares_generator))
````

### List comprehension vs generator expression

````python
numbers = [1, 2, 3, 4]

list_result = [x ** 2 for x in numbers]
generator_result = (x ** 2 for x in numbers)

print(list_result)
print(generator_result)
print(list(generator_result))
````

### Practical example: summing large data

A generator expression can avoid constructing an intermediate list when only an aggregate result is needed.

````python
numbers = range(1, 1_000_001)

total = sum(x * x for x in numbers)

print(total)
````

## 16. Generator Functions with `yield`

A generator function uses `yield` to produce values lazily.

Calling the function returns a generator rather than executing the entire function immediately.

````python
def count_up_to(limit):
    number = 1

    while number <= limit:
        yield number
        number += 1

counter = count_up_to(3)

print(counter)
print(next(counter))
print(next(counter))
print(next(counter))
````

### Using the generator in a loop

````python
def squares(limit):
    for number in range(1, limit + 1):
        yield number ** 2

for value in squares(5):
    print(value)
````

### Edge cases

Once a generator is exhausted, further `next()` calls raise `StopIteration`.

````python
generator = (x for x in [1, 2])

print(list(generator))

# next(generator)  # StopIteration
````

## 17. `map()` and Generator Expressions

Both can provide lazy transformations, but generator expressions often read more naturally for straightforward expressions.

````python
numbers = [1, 2, 3, 4]

mapped = map(lambda x: x * 2, numbers)
generated = (x * 2 for x in numbers)

print(list(mapped))
print(list(generated))
````

### Practical guideline

Use whichever form makes the transformation clearer. `map()` can be particularly useful when you already have a named function or when expressing a functional pipeline.

## 18. `filter()` and Generator Expressions

Filtering can also be expressed with a generator expression.

````python
numbers = range(10)

filtered = filter(lambda x: x % 2 == 0, numbers)
generated = (x for x in numbers if x % 2 == 0)

print(list(filtered))
print(list(generated))
````

### Practical guideline

A generator expression is often concise when the filtering condition and transformation are simple.

## 19. `reduce()` vs Built-in Aggregations

`reduce()` is flexible, but Python provides specialized built-ins for many common reductions.

Examples:
- `sum()`
- `min()`
- `max()`
- `any()`
- `all()`
- `len()`

````python
from functools import reduce

numbers = [1, 2, 3, 4]

reduce_sum = reduce(lambda a, b: a + b, numbers)
builtin_sum = sum(numbers)

print(reduce_sum)
print(builtin_sum)
````

### Practical guideline

Prefer specialized built-ins when they express the operation directly. Use `reduce()` when the accumulation logic is not already represented clearly by a standard built-in.

````python
numbers = [3, 8, 2, 10]

print(sum(numbers))
print(min(numbers))
print(max(numbers))
print(any(x > 9 for x in numbers))
print(all(x > 0 for x in numbers))
````

## 20. Iterator Consumption

Iterators maintain state. Once values have been consumed, they are not automatically reset.

````python
numbers = [1, 2, 3]

iterator = iter(numbers)

print(next(iterator))
print(list(iterator))

# The iterator is now exhausted.
print(list(iterator))
````

### Reusable iterable vs single-use iterator

A list can create a fresh iterator each time. An iterator itself continues from its current position.

````python
numbers = [1, 2, 3]

print(list(numbers))
print(list(numbers))

iterator = iter(numbers)

print(list(iterator))
print(list(iterator))
````

## 21. `iter()` with a Sentinel

`iter(callable, sentinel)` creates an iterator that repeatedly calls a zero-argument callable until the returned value equals the sentinel.

This is useful for repeatedly reading values until a terminating value appears.

````python
values = iter([10, 20, 30, "STOP", 40])

# next() can still consume the underlying iterator.
print(next(values))

# A practical sentinel pattern can be used with a callable.
data = iter([10, 20, 30, None])

read_value = lambda: next(data)

for value in iter(read_value, None):
    print(value)
````

### Important point

The sentinel form is:

```python
iter(callable, sentinel)
```

The callable is repeatedly invoked, and iteration stops when its returned value equals the sentinel.

## 22. Combining Lazy Operations

`map()` and `filter()` can be chained without immediately creating intermediate lists.

````python
numbers = range(1, 11)

even_numbers = filter(lambda x: x % 2 == 0, numbers)
squared_even_numbers = map(lambda x: x ** 2, even_numbers)

for value in squared_even_numbers:
    print(value)
````

### Pipeline with a final aggregation

````python
from functools import reduce

numbers = range(1, 11)

result = reduce(
    lambda a, b: a + b,
    map(
        lambda x: x ** 2,
        filter(lambda x: x % 2 == 0, numbers)
    )
)

print(result)
````

# Combined Example: Data Processing Pipeline

This example demonstrates a complete transformation pipeline:

1. Start with raw values.
2. Filter valid values.
3. Transform the values.
4. Aggregate the result.

````python
from functools import reduce

raw_scores = [85, -10, 72, 101, 90, 64]

# Keep scores within the valid range
valid_scores = filter(lambda score: 0 <= score <= 100, raw_scores)

# Convert each score to a percentage of 1.0
normalized_scores = map(lambda score: score / 100, valid_scores)

# Add all normalized scores
total = reduce(lambda a, b: a + b, normalized_scores, 0)

print("Total normalized score:", total)
````

# Quick Reference

| Tool | Purpose | Typical result |
|---|---|---|
| `map(f, iterable)` | Transform elements | Iterator |
| `map(f, a, b)` | Transform multiple iterables | Iterator |
| `filter(f, iterable)` | Keep matching elements | Iterator |
| `filter(None, iterable)` | Keep truthy elements | Iterator |
| `reduce(f, iterable)` | Accumulate values | Single value |
| `reduce(f, iterable, initial)` | Accumulate with initial value | Single value |
| `iter(iterable)` | Create iterator | Iterator |
| `next(iterator)` | Get next item | One value |
| `next(iterator, default)` | Get next item safely | Value/default |
| Generator expression | Lazy transformation | Generator |
| Generator function | Lazy production with `yield` | Generator |
| `list(iterator)` | Materialize iterator | List |

# Common Pitfalls

- `map()` and `filter()` return iterators in Python 3, not lists.
- Iterators are generally single-pass and are consumed as they are iterated.
- `next()` on an exhausted iterator raises `StopIteration` unless a default is supplied.
- `map()` with multiple iterables stops at the shortest iterable.
- `reduce()` on an empty iterable requires an initializer.
- `reduce()` is not always the clearest choice; use built-in aggregations when they express the operation better.
- Generator expressions are lazy and do not immediately create a list.
- A list is iterable but is not itself an iterator.
- `iter()` converts an iterable into an iterator.
- `iter(callable, sentinel)` repeatedly calls the callable until the sentinel is returned.
- Materialize lazy results with `list()`, `tuple()`, or another appropriate collection when repeated access is required.
