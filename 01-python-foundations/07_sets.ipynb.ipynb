# Sets

A reference notebook covering Python sets, `frozenset`, set operations, comprehensions, hashability, collection access patterns, and conversions between lists, sets, and dictionaries.

## 1. Creating Sets

A `set` is an unordered, mutable collection of unique, hashable elements. Use `{}` for a non-empty set and `set()` for an empty set.

```python
numbers = {1, 2, 3, 4}
names = {"Alice", "Bob", "Charlie"}
empty_set = set()
empty_dict = {}
print(numbers)
print(names)
print(type(empty_set))
print(type(empty_dict))
```

### Edge cases

- Set elements must be hashable.
- Sets do not provide positional indexing.
- Duplicate elements are automatically removed.
- Do not rely on iteration order for positional behavior.

```python
values = {1, 2, 2, 3, 3}
print(values)
# {[1, 2], [3, 4]}  # TypeError: lists are unhashable
```

## 2. Accessing Lists, Dictionaries, and Sets

Lists use positional indexes and slices; dictionaries use keys; sets do not support positional access and are normally queried with membership testing or iteration.

```python
items = ["Python", "NumPy", "Pandas"]
print(items[0])
print(items[-1])
print(items[0:2])

student = {"name": "Aarav", "age": 21, "course": "CSE"}
print(student["name"])
print(student.get("age"))

skills = {"Python", "SQL", "Git"}
print("Python" in skills)
for skill in skills:
    print(skill)
# skills[0]  # TypeError: sets are not subscriptable
```

### Converting a set for positional access

If an API requires a sequence, convert the set to a list first. The resulting order should not be treated as a meaningful original set order.

```python
skills = {"Python", "SQL", "Git"}
skills_list = list(skills)
print(skills_list)
print(skills_list[0])
```

## 3. `add()` and `update()`

```python
skills = {"Python", "SQL"}
skills.add("Git")
print(skills)
skills.update(["Docker", "Linux"])
print(skills)
skills.update({"GitHub", "AWS"})
print(skills)
```

`add()` inserts one element. `update()` adds elements from one or more iterables. Updating with a string adds its characters individually.

```python
values = {1, 2}
values.add(2)
values.update("AI")
print(values)
```

## 4. `remove()` vs `discard()`

`remove(x)` raises `KeyError` when absent; `discard(x)` silently does nothing when absent.

```python
numbers = {1, 2, 3, 4}
numbers.remove(3)
numbers.discard(4)
numbers.discard(100)
print(numbers)
# numbers.remove(100)  # KeyError
```

## 5. `pop()` and `clear()`

`pop()` removes and returns an arbitrary element. `clear()` removes everything.

```python
values = {10, 20, 30, 40}
removed = values.pop()
print("Removed:", removed)
print("Remaining:", values)
values.clear()
print(values)
# set().pop()  # KeyError
```

## 6. `copy()`

`copy()` creates a shallow copy of a set.

```python
original = {1, 2, 3}
copied = original.copy()
copied.add(4)
print(original)
print(copied)
print(original is copied)
```

## 7. `frozenset`

`frozenset` is the immutable, hashable counterpart of `set`. It can be used as a dictionary key or as an element of another set.

```python
permissions = frozenset({"read", "write"})
print(permissions)
print("read" in permissions)
# permissions.add("execute")  # AttributeError

data = {permissions: "Admin permissions"}
print(data[permissions])
```

## 8. Set Comprehension

Syntax: `{expression for item in iterable if condition}`. Duplicate results are automatically collapsed.

```python
numbers = {1, 2, 3, 4, 5, 6}
squares = {x ** 2 for x in numbers}
even_numbers = {x for x in numbers if x % 2 == 0}
print(squares)
print(even_numbers)

words = ["Python", "python", "AI", "Python"]
print({word.lower() for word in words})
```

## 9. Mathematical Set Operators

`|` union, `&` intersection, `-` difference, `^` symmetric difference.

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print("Union:", a | b)
print("Intersection:", a & b)
print("Difference:", a - b)
print("Symmetric difference:", a ^ b)
```

### Practical example

```python
python_students = {"alice", "bob", "charlie"}
ml_students = {"bob", "charlie", "david"}
print("All:", python_students | ml_students)
print("Both:", python_students & ml_students)
print("Only Python:", python_students - ml_students)
print("Only one:", python_students ^ ml_students)
```

## 10. Set Methods and Relationship Checks

Method forms include `union()`, `intersection()`, `difference()`, `symmetric_difference()`. Relationship checks include `issubset()`, `issuperset()`, and `isdisjoint()`.

```python
a = {1, 2, 3}
b = {3, 4, 5}
print(a.union(b))
print(a.intersection(b))
print(a.difference(b))
print(a.symmetric_difference(b))

small = {1, 2}
large = {1, 2, 3, 4}
print(small.issubset(large))
print(large.issuperset(small))
print({1, 2}.isdisjoint({3, 4}))
```

## 11. In-place Set Operations

Methods ending in `_update` modify the existing set.

```python
a = {1, 2, 3}
b = {3, 4, 5}
a.intersection_update(b)
print(a)
a = {1, 2, 3}
a.difference_update(b)
print(a)
```

## 12. Hashable vs Unhashable Objects

Set elements must be hashable. Typical hashable values include integers, strings, tuples containing hashable values, and `frozenset`. Lists, dictionaries, and sets are unhashable.

```python
valid = {1, "Python", (1, 2), frozenset({3, 4})}
print(valid)
print(hash(42))
print(hash("Python"))
print(hash((1, 2)))
# hash([1, 2])  # TypeError
```

## 13. Lists to Sets

Use `set(list_value)` to remove duplicates and obtain a set. This loses list-style positional structure.

```python
items = ["b", "a", "b", "c", "a"]
unique_items = set(items)
print(unique_items)

# Preserve first-seen order while removing duplicates:
unique_in_order = list(dict.fromkeys(items))
print(unique_in_order)
```

## 14. Sets to Lists

Use `list(set_value)` when a sequence is required. The resulting order should not be treated as meaningful set order.

```python
skills = {"Python", "SQL", "Git"}
skills_list = list(skills)
print(skills_list)
print(skills_list[0])
```

## 15. Lists to Dictionaries

A list of 2-item pairs can be passed to `dict()`. Two parallel lists can be combined with `zip()` first.

```python
pairs = [("name", "Alice"), ("age", 21), ("course", "CSE")]
student = dict(pairs)
print(student)
print(student["name"])

keys = ["name", "age", "course"]
values = ["Alice", 21, "CSE"]
print(dict(zip(keys, values)))
```

### Edge case: unequal lengths

`zip()` stops at the shortest iterable.

```python
keys = ["a", "b", "c"]
values = [10, 20]
print(dict(zip(keys, values)))
```

## 16. Dictionaries to Lists

Use `list(d)` for keys, `list(d.values())` for values, and `list(d.items())` for key-value pairs.

```python
student = {"name": "Alice", "age": 21, "course": "CSE"}
print(list(student))
print(list(student.values()))
items = list(student.items())
print(items)
print(items[0][0], items[0][1])
```

## 17. Dictionaries to Sets

`set(d)` creates a set of dictionary keys. Use `set(d.values())` for values and `set(d.items())` for key-value pairs when the values are hashable.

```python
student = {"name": "Alice", "age": 21, "course": "CSE"}
print(set(student))
print(set(student.values()))
print(set(student.items()))

data = {"a": 10, "b": 20, "c": 10}
print(set(data.values()))
```

## 18. Sets to Dictionaries

A set has no inherent key-value relationship. You can use its elements as keys with `dict.fromkeys()`, or transform elements into key-value pairs with a comprehension.

```python
keys = {"python", "sql", "git"}
print(dict.fromkeys(keys, True))

scores = {80, 90, 95}
score_labels = {score: f"score_{score}" for score in scores}
print(score_labels)
```

## 19. Collection Conversion Map

Common conversions: `list -> set` with `set(x)`; `set -> list` with `list(x)`; list of pairs -> dictionary with `dict(x)`; keys + values -> dictionary with `dict(zip(keys, values))`; dictionary -> keys/values/items with `list(d)`, `list(d.values())`, `list(d.items())`; dictionary -> key set with `set(d)`; set -> dictionary with `dict.fromkeys(s, value)`.

```python
items = ["Python", "SQL", "Python", "Git"]
unique_items = set(items)
items_again = list(unique_items)
item_flags = dict.fromkeys(items, True)
key_set = set(item_flags)
pairs = list(item_flags.items())
print("List:", items)
print("Set:", unique_items)
print("List again:", items_again)
print("Dictionary:", item_flags)
print("Keys as set:", key_set)
print("Items:", pairs)
```

## 20. Membership Testing

`in` behaves differently by collection: list membership searches elements, set membership uses hash-based lookup, and dictionary membership checks keys by default.

```python
items = ["Python", "SQL", "Git"]
skills = {"Python", "SQL", "Git"}
student = {"name": "Alice", "age": 21}
print("Python" in items)
print("Python" in skills)
print("name" in student)
print("Alice" in student.values())
```

## Combined Example: Data Cleaning

Sets are useful for deduplicating categories; dictionaries can map the cleaned values; lists can preserve a sequence for downstream code.

```python
raw_categories = ["Python", "Machine Learning", "Python", "Data Science", "Machine Learning"]
unique_categories = set(raw_categories)
category_codes = {category: index for index, category in enumerate(sorted(unique_categories), start=1)}
categories = list(category_codes)
print("Unique:", unique_categories)
print("Codes:", category_codes)
print("Ordered categories:", categories)
print("Python" in unique_categories)
```

# Common Pitfalls

- Sets do not support indexing or slicing.
- Do not rely on set iteration order for positional behavior.
- `discard()` ignores missing elements; `remove()` raises `KeyError`.
- `pop()` removes an arbitrary element from a set.
- Set elements must be hashable.
- List -> set removes duplicates and positional structure.
- Set -> list gives positions, but not a meaningful original order.
- `set(dictionary)` produces dictionary keys, not values.
- `dict(zip(keys, values))` stops at the shortest input.
- Dictionary keys are unique; duplicate keys overwrite earlier values.
- A list is accessed by position, a dictionary by key, and a set primarily by membership testing.
