## 1. String Basics

A Python `str` is an immutable sequence of Unicode characters. Strings can be created with single quotes, double quotes, or triple quotes.

````python
name = "Python"
message = 'Hello, Python!'
multiline = """This is
a multiline string."""

print(name)
print(message)
print(multiline)
````

## 2. String Indexing

Strings support zero-based indexing.

- Positive indexes start at `0`.
- Negative indexes start at `-1` from the end.

````python
text = "Python"

print(text[0])    # P
print(text[1])    # y
print(text[-1])   # n
print(text[-2])   # o
````

## 3. String Slicing

General syntax:

```python
string[start:stop:step]
```

`stop` is exclusive. `start` and `stop` can be omitted.

````python
text = "Python"

print(text[0:2])    # Py
print(text[2:5])    # tho
print(text[:4])     # Pyth
print(text[2:])     # thon
print(text[:])      # Python
print(text[::2])    # Pto
print(text[1::2])   # yhn
print(text[::-1])   # nohtyP
````

## 4. Slicing Edge Cases

Slicing is forgiving: out-of-range boundaries normally do not raise `IndexError`.

A slice with a step of zero is invalid.

````python
text = "Python"

print(text[100:200])  # ''
print(text[-100:3])   # 'Pyt'
print(text[4:1])      # ''
print(text[::-1])     # reversed

# This raises ValueError:
# print(text[::0])
````

## 5. String Methods: Case Conversion

````python
text = "hello Python"

print(text.lower())
print(text.upper())
print(text.capitalize())
print(text.title())
print(text.swapcase())
````

## 6. String Methods: Removing Whitespace

`strip()`, `lstrip()`, and `rstrip()` remove characters from the ends of a string.

Without arguments, they remove common whitespace.

````python
text = "   Python   "

print(repr(text.strip()))
print(repr(text.lstrip()))
print(repr(text.rstrip()))

text2 = "---Python---"
print(text2.strip("-"))
````

## 7. Searching Strings

- `find()` returns `-1` when the substring is not found.
- `index()` raises `ValueError` when it is not found.
- `startswith()` and `endswith()` test prefixes and suffixes.

````python
text = "machine learning"

print(text.find("learn"))
print(text.find("xyz"))

print(text.index("learn"))

print(text.startswith("machine"))
print(text.endswith("learning"))

# text.index("xyz")  # ValueError
````

## 8. Counting and Replacing

````python
text = "banana"

print(text.count("a"))
print(text.count("an"))

print(text.replace("banana", "apple"))
print(text.replace("a", "A"))
print(text.replace("a", "A", 1))
````

## 9. Splitting Strings

`split()` converts a string into a list.

`splitlines()` separates lines in multiline text.

````python
text = "Python,Java,C++"

print(text.split(","))
print("one two three".split())

multiline = "line one\nline two\nline three"
print(multiline.splitlines())
````

## 10. `partition()`

`partition(separator)` always returns a 3-item tuple:

```text
(before, separator, after)
```

If the separator is not found, the separator and after-part are empty.

````python
text = "name=Jayaadithya"

print(text.partition("="))
print("hello".partition("="))
````

## 11. `join()`

`join()` combines an iterable of strings using the calling string as the separator.

````python
words = ["Python", "is", "useful"]

print(" ".join(words))
print("-".join(words))
print("".join(["A", "B", "C"]))

# Every element must be a string.
# ",".join([1, 2, 3])  # TypeError
````

## 12. String Classification Methods

Useful methods include:

- `isdigit()`
- `isnumeric()`
- `isalpha()`
- `isalnum()`
- `islower()`
- `isupper()`
- `isspace()`
- `istitle()`

````python
samples = ["123", "²", "Python", "Python3", "hello", "HELLO", "   ", "Hello World"]

for value in samples:
    print(
        value,
        value.isdigit(),
        value.isnumeric(),
        value.isalpha(),
        value.isalnum(),
        value.islower(),
        value.isupper(),
        value.isspace(),
        value.istitle(),
    )
````

## 13. Removing Prefixes and Suffixes

`removeprefix()` and `removesuffix()` remove an exact prefix/suffix when present.

Unlike `strip()`, they do not repeatedly remove individual characters.

````python
filename = "prefix_data.csv"

print(filename.removeprefix("prefix_"))
print(filename.removesuffix(".csv"))

print("---Python---".strip("-"))
print("---Python---".removeprefix("---"))
````

## 14. String Formatting with f-Strings

F-strings are the preferred modern formatting approach for most Python code.

````python
name = "Jayaadithya"
score = 95

print(f"Name: {name}")
print(f"{name} scored {score} marks")
print(f"Double score: {score * 2}")
````

## 15. Number Formatting with f-Strings

Format specifications can control precision, separators, width, alignment, and percentages.

````python
value = 1234567.89123

print(f"{value:,.2f}")
print(f"{value:.3f}")
print(f"{0.875:.1%}")

name = "Python"
print(f"{name:>10}")
print(f"{name:<10}")
print(f"{name:^10}")
print(f"{name:*^10}")
````

## 16. `str.format()`

`str.format()` is an older but still useful formatting mechanism.

````python
name = "Python"
version = 3.14

print("Language: {}, Version: {}".format(name, version))
print("Language: {language}, Version: {version}".format(
    language=name,
    version=version
))
````

## 17. Old `%` Formatting

Percent-style formatting is older Python syntax. It still exists in Python codebases, but f-strings are generally clearer for new code.

````python
name = "Python"
version = 3

print("Language: %s, Version: %d" % (name, version))
print("Accuracy: %.2f%%" % 97.456)
````

## 18. Escaping Characters

Backslash escapes represent characters that would otherwise have special meaning.

Common examples:
- `\n` newline
- `\t` tab
- `\\` backslash
- `\"` double quote
- `\'` single quote

````python
print("Line 1\nLine 2")
print("Column1\tColumn2")
print("A backslash: \\")

print("He said, \"Hello\"")
print('It\'s Python')
````

## 19. Raw Strings

Raw strings treat backslashes mostly literally. They are especially useful for regular expressions and Windows-style paths.

A raw string cannot end with a single backslash.

````python
path = r"C:\Users\Jayaadithya\Documents"
pattern = r"\d+\.\d+"

print(path)
print(pattern)
````

## 20. String Immutability

Strings cannot be modified in place.

Operations such as `replace()`, `upper()`, and concatenation create new string objects rather than modifying the original string.

````python
text = "python"

# text[0] = "P"  # TypeError

text = text.upper()
print(text)

original = "python"
changed = original.replace("p", "P")

print(original)
print(changed)
````

## 21. Equality vs Identity for Strings

Use `==` to compare string values.

Use `is` for object identity; it should not be used for ordinary string-value comparisons. String interning can make some equal strings share identity, but that behavior should not be relied upon.

````python
a = "python"
b = "python"

print(a == b)  # value comparison
print(a is b)  # identity comparison; may be True due to interning

x = "".join(["py", "thon"])
print(x == a)  # True
print(x is a)  # identity is not something to rely on
````

## 22. String Concatenation

Strings can be concatenated with `+`. For many repeated concatenations, building a list and using `join()` is often preferable.

````python
first = "Hello"
second = "World"

print(first + " " + second)

parts = ["Python", "data", "science"]
result = " ".join(parts)
print(result)
````

## 23. Iterating Over a String

A string is iterable, so a `for` loop visits one character at a time.

````python
text = "AI"

for character in text:
    print(character)
````

# Tuples

A tuple is an ordered, immutable sequence. Tuples are useful for fixed collections of values and for returning multiple values from functions.

## 24. Creating Tuples

````python
empty = ()
numbers = (1, 2, 3)
mixed = ("Python", 3, True)

print(empty)
print(numbers)
print(mixed)

single = (42,)
not_a_tuple = (42)

print(type(single))
print(type(not_a_tuple))
````

## 25. Tuple Indexing and Slicing

Tuples support the same indexing and slicing style as other sequences.

````python
data = ("Python", "Java", "C++", "Go")

print(data[0])
print(data[-1])
print(data[1:3])
print(data[::-1])
````

## 26. Tuple Methods

Because tuples are immutable, they have fewer methods than lists.

The main tuple methods are `count()` and `index()`.

````python
values = (1, 2, 2, 3, 2)

print(values.count(2))
print(values.index(3))
````

## 27. Tuple Immutability

Tuple elements cannot be reassigned.

However, a tuple can contain a mutable object such as a list. The tuple cannot replace the list itself, but the list's contents can still change.

````python
values = (1, 2, 3)

# values[0] = 10  # TypeError

nested = ("Python", [1, 2, 3])
nested[1].append(4)

print(nested)
````

## 28. Tuples as Dictionary Keys

Tuples can be dictionary keys when all their elements are hashable.

This is useful for representing coordinates or compound keys.

````python
locations = {
    (0, 0): "origin",
    (10, 20): "point A",
}

print(locations[(10, 20)])

# This would fail because lists are unhashable:
# data = {[1, 2]: "value"}
````

## 29. Tuple vs List

Use a list when the collection is expected to change.

Use a tuple when the collection represents a fixed sequence of values or when immutability is useful.

````python
items = ["Python", "Java"]
items.append("C++")

fixed = ("Python", "Java")

print(items)
print(fixed)
````

## 30. Tuple Packing

Packing means placing multiple values into a tuple without explicitly writing parentheses.

````python
packed = 10, 20, 30

print(packed)
print(type(packed))
````

## 31. Tuple Unpacking

Unpacking assigns tuple elements to multiple variables.

The number of variables normally has to match the number of values.

````python
point = (10, 20)

x, y = point

print(x)
print(y)

# a, b = (1, 2, 3)  # ValueError
````

## 32. Extended Unpacking with `*`

The starred target collects zero or more remaining values into a list.

````python
values = (1, 2, 3, 4, 5)

first, *middle, last = values

print(first)
print(middle)
print(last)

*start, last = values
print(start, last)
````

## 33. Swapping Variables

Python can swap variables without a temporary variable by using tuple packing and unpacking.

````python
a = 10
b = 20

a, b = b, a

print(a)
print(b)
````

## 34. Unpacking in Loops

Sequence unpacking is useful when iterating over pairs or tuples.

````python
records = [
    ("Python", 3),
    ("Java", 21),
    ("C++", 23),
]

for language, version in records:
    print(f"{language}: {version}")
````

## 35. Multiple Return Values

Python functions technically return one object, but that object can be a tuple containing multiple values.

````python
def min_max(values):
    return min(values), max(values)

result = min_max([4, 1, 9, 2])

print(result)
low, high = result

print(low)
print(high)
````

## 36. Practical Example: Model Configuration

Tuples and formatted strings can be useful for representing fixed configuration values and generating readable summaries.

````python
model_config = ("RandomForest", 200, "gini")

name, estimators, criterion = model_config

summary = (
    f"Model={name}, "
    f"estimators={estimators}, "
    f"criterion={criterion}"
)

print(summary)
````

## 37. Practical Example: Parsing a Record

String methods can be combined with tuple unpacking to parse structured text.

````python
record = "Jayaadithya,Python,95"

name, language, score_text = record.split(",")

score = int(score_text)

print(name)
print(language)
print(score)
````

## 38. Practical Example: File Extension

`rsplit()` can separate a filename from its extension when only the final separator is relevant.

````python
filename = "model.v2.final.pkl"

name, extension = filename.rsplit(".", 1)

print(name)
print(extension)
````

## 39. Common String Edge Cases

### Empty strings
An empty string is falsy.

### Indexing
Indexing outside the valid range raises `IndexError`.

### Slicing
Out-of-range slice boundaries are normally safe.

### Zero slice step
A step of `0` raises `ValueError`.

### `join()`
All elements must be strings.

### `index()`
Missing values raise `ValueError`; `find()` returns `-1`.

````python
empty = ""

print(bool(empty))

# print(empty[0])  # IndexError

print(empty[:10])  # ''

# print("abc"[::0])  # ValueError

# ",".join([1, 2])  # TypeError

print("abc".find("z"))  # -1
# print("abc".index("z"))  # ValueError
````

## 40. Common Tuple Edge Cases

- A one-element tuple requires a trailing comma.
- Unpacking requires compatible numbers of values unless `*` is used.
- Tuples are immutable, but objects stored inside them may be mutable.
- A tuple is hashable only when all of its elements are hashable.

````python
single = (10,)
print(type(single))

values = (1, [2, 3])
values[1].append(4)
print(values)

hashable_tuple = (1, "Python")
print(hash(hashable_tuple))

# unhashable_tuple = (1, [2, 3])
# hash(unhashable_tuple)  # TypeError
````

## Common Pitfalls

1. Strings and tuples are immutable; methods return new values rather than modifying them.
2. Use `==` for string value comparison, not `is`.
3. A one-element tuple needs a comma: `(value,)`.
4. `find()` and `index()` behave differently when a substring is missing.
5. `strip()` removes characters from the ends; it does not remove an exact substring.
6. `join()` expects an iterable of strings.
7. Tuple immutability does not make nested mutable objects immutable.
8. Slice boundaries can be out of range without raising an error, but direct indexing cannot.
